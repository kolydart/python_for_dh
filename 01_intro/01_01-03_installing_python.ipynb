## Εγκατάσταση της Python τοπικά

Το Anaconda Navigator είναι μια φιλική προς τον χρήστη γραφική διεπαφή που αναλαμβάνει την εγκατάσταση για λογαριασμό σας. Σας επιτρέπει επίσης να δημιουργείτε περιβάλλοντα (environments), δηλαδή απομονωμένους χώρους στον υπολογιστή σας, καθέναν με τη δική του έκδοση της Python και τις δικές του εγκατεστημένες βιβλιοθήκες. Σε αυτό το κεφάλαιο θα σας καθοδηγήσω στα βήματα εγκατάστασης του Anaconda Navigator στον υπολογιστή σας, ανεξάρτητα από το λειτουργικό σύστημα.

Κώδικα σε Python γράφουμε με διάφορους τρόπους, ανάλογα με τη χρήση για την οποία προορίζεται. Άλλοτε, πάλι, θα χρησιμοποιήσετε την Python στο τερματικό, για να εκτελέσετε γρήγορα κάποιες εργασίες σε δεδομένα ενός καταλόγου.
