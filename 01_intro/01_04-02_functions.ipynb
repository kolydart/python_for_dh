# Συναρτήσεις

## Εισαγωγή

Σε αυτή την ενότητα θα μάθετε όλα τα βασικά για τις συναρτήσεις. Έχετε ήδη συναντήσει αρκετές συναρτήσεις στο εγχειρίδιο, όπως την `print()`. Στόχος της ενότητας είναι να μάθετε να δημιουργείτε τις δικές σας.

Οι **συναρτήσεις** (functions) υπάρχουν σε όλες τις γλώσσες προγραμματισμού. Σας επιτρέπουν να γράψετε ένα μπλοκ κώδικα που μπορεί να χρησιμοποιηθεί αργότερα στο πρόγραμμά σας. Είναι απαραίτητες, επειδή σας απαλλάσσουν από το να γράφετε το ίδιο μπλοκ κώδικα ξανά και ξανά. Καλός κώδικας είναι ο κώδικας που δεν επαναλαμβάνεται. Με άλλα λόγια, αν το πρόγραμμά σας πρέπει να εκτελεί επανειλημμένα μια εργασία, ο κώδικας για την εργασία αυτή πρέπει να οργανωθεί σε μια καθαρή συνάρτηση (ή σε μια κλάση, την οποία θα γνωρίσουμε στην επόμενη ενότητα).

Όπως θα δούμε, μια συνάρτηση έχει πέντε μέρη:

1. το όνομα της συνάρτησης
2. τα ορίσματα που δίνονται στη συνάρτηση (προαιρετικά)
3. ένα docstring που εξηγεί τη συνάρτηση (προαιρετικό)
4. τον κώδικα της συνάρτησης
5. τα δεδομένα που επιστρέφει (προαιρετικά)

Στο τέλος της ενότητας θα έχετε κατανοήσει καθένα από αυτά τα στοιχεία.

## Οι συναρτήσεις στην πράξη

Πιστεύω ότι ο καλύτερος τρόπος για να μάθετε τις συναρτήσεις είναι να τις δείτε στην πράξη, πριν αναλύσουμε τι συμβαίνει. Ας δούμε τον παρακάτω κώδικα.

````python
def adding_one(number):
    x = number+1
    return x
````

Ας αναλύσουμε τον παραπάνω κώδικα ξεκινώντας από την πρώτη γραμμή.
```
def adding_one(number):
```

Εδώ χρησιμοποιούμε τη λέξη `def`. Στην Python, το `def` προέρχεται από το define (ορίζω), επειδή ορίζουμε μια συνάρτηση. Όταν η Python συναντά το `def`, περιμένει ότι ό,τι ακολουθεί θα τηρεί τη σύνταξη μιας συνάρτησης.

Το επόμενο στοιχείο της πρώτης γραμμής είναι το `adding_one`, το όνομα της συνάρτησης (όπως αναμένεται, το όνομα της συνάρτησης `print` είναι `print`). Με αυτό το όνομα θα καλείτε τη συνάρτησή σας αργότερα στο σενάριό σας.

Ακολουθεί το `(number)`. Το όνομα μιας συνάρτησης Python πρέπει να ακολουθείται από παρενθέσεις. Μέσα στις παρενθέσεις μπορείτε να τοποθετήσετε μία ή περισσότερες παραμέτρους (parameters), οι οποίες είναι προαιρετικές. Οι παράμετροι σας επιτρέπουν να δίνετε ορίσματα (arguments) σε μια συνάρτηση. Η συνάρτηση χρησιμοποιεί αυτά τα ορίσματα για να εκτελέσει κάποια ενέργεια, είτε τροποποιώντας την παράμετρο είτε αξιοποιώντας την με κάποιον άλλο τρόπο. Στη δική μας συνάρτηση, η μόνη παράμετρος είναι η `number`. Αν ορίσετε περισσότερες παραμέτρους, πρέπει να τις χωρίσετε με κόμμα, όπως θα δούμε παρακάτω. Παρατηρήστε ότι οι παράμετροι δεν περικλείονται σε εισαγωγικά, επειδή δεν είναι συμβολοσειρές, αλλά παράμετροι που λειτουργούν ως μεταβλητές μέσα στον κώδικα της συνάρτησης.

Τέλος, η πρώτη γραμμή κλείνει με `:`. Αυτό δηλώνει στην Python δύο πράγματα: πρώτον, ότι η σύνταξη του ορισμού της συνάρτησης ολοκληρώθηκε· δεύτερον, ότι οι επόμενες γραμμές, με εσοχή, περιέχουν τον κώδικα της συνάρτησης. Η `:` και η εσοχή είναι υποχρεωτικές.

Η επόμενη γραμμή κώδικα είναι:
```
x = number+1
```

Αυτός είναι ο κώδικας της συνάρτησης. Εδώ δημιουργούμε μέσα στη συνάρτηση μια μεταβλητή με το όνομα `x`, η οποία ισούται με το άθροισμα του `number` που δίνεται στη συνάρτηση και του `1`.

Το τελευταίο τμήμα του κελιού είναι

```
return x
```

Η εντολή αυτή επιστρέφει στον χρήστη το `x` που δημιουργήσαμε προσωρινά μέσα στη συνάρτηση.

Ας καλέσουμε τώρα τη συνάρτηση στο παρακάτω κελί για να δούμε πώς λειτουργεί.

````python
result = adding_one(1)
````

Στον παραπάνω κώδικα δημιουργήσαμε ένα νέο αντικείμενο, το `result`, το οποίο θα περιέχει το αποτέλεσμα της συνάρτησης. Όπως κάθε αντικείμενο, μπορεί να έχει όποιο όνομα θέλετε, εκτός από τις δεσμευμένες λέξεις της Python.

Στη συνέχεια καλούμε τη συνάρτηση `adding_one` και της δίνουμε ένα μόνο όρισμα, τον ακέραιο `1`.

## Συναρτήσεις με πολλά ορίσματα

Παραπάνω είδαμε μια απλή συνάρτηση με ένα μόνο όρισμα. Στην Python μπορείτε να ορίσετε σε μια συνάρτηση όσα ορίσματα θέλετε. Ας φτιάξουμε μια ελαφρώς διαφορετική συνάρτηση, που προσθέτει δύο αριθμούς τους οποίους δίνει ο χρήστης.

````python
def my_function2(number1, number2):
    x = number1+number2
    return x
````

Παρατηρήστε ότι η `my_function2` είναι ίδια με την παραπάνω συνάρτηση, με τη διαφορά ότι έχει δύο ορίσματα, `number1` και `number2`. Επίσης, μέσα στη συνάρτηση το `x` είναι το άθροισμα του `number1` και του `number2`.

Ας χρησιμοποιήσουμε τώρα τη συνάρτηση, δίνοντάς της δύο αριθμούς: 1 και 3.

````python
result2 = my_function2(1,3)
print (result2)
````

Ωραία! Πήραμε το αποτέλεσμα που θέλαμε. Τα ορίσματα αυτά ονομάζονται θεσιακά ορίσματα (positional arguments), επειδή η αντιστοίχισή τους εξαρτάται από τη θέση τους μέσα στις παρενθέσεις της συνάρτησης. Ας τροποποιήσουμε ελαφρώς τη συνάρτηση, για να δείτε τι εννοώ.

````python
def my_function3(number1, number2):
    print ("Number 1 is ", number1)
    print ("Number 2 is ", number2)
````

````python
my_function3(1, 3)
````

Παρατηρήστε ότι αφαίρεσα τη γραμμή `return`, επειδή αυτή η συνάρτηση δεν χρειάζεται να επιστρέψει τίποτα στον χρήστη· ο μόνος της σκοπός είναι να εκτυπώσει τα δύο ορίσματα. Ας αντιστρέψουμε τώρα τη σειρά των ορισμάτων.

````python
my_function3(3, 1)
````

Επειδή πρόκειται για θεσιακά ορίσματα, η σωστή αντιστοίχισή τους εξαρτάται από τη θέση τους. Μπορούμε να το παρακάμψουμε ορίζοντας ρητά σε ποια παράμετρο αντιστοιχεί κάθε τιμή, ώστε να μην εξαρτόμαστε από τη σειρά με την οποία δίνουμε τα ορίσματα στη συνάρτηση. Δείτε το στην πράξη στον παρακάτω κώδικα.

````python
my_function3(number2=3, number1=1)
````

## Ορίσματα λέξης-κλειδιού (keyword arguments)

Μερικές φορές, όταν δημιουργούμε μια συνάρτηση, θέλουμε ένα όρισμα να είναι προαιρετικό για τον χρήστη. Σε αυτές τις περιπτώσεις ορίζουμε ένα όρισμα λέξης-κλειδιού (keyword argument), δηλαδή μια παράμετρο με προεπιλεγμένη τιμή. Στην παρακάτω συνάρτηση θέλουμε να δώσουμε στον χρήστη τη δυνατότητα να ορίσει ένα επώνυμο. Παρατηρήστε, ωστόσο, ότι η προεπιλεγμένη τιμή είναι "Mattingly".

````python
def add_surname(first_name, last_name = "Mattingly"):
    print (first_name, last_name)
````

````python
add_surname("William")
````

Η συνάρτηση λειτούργησε σωστά, αλλά ο χρήστης μπορεί ακόμη να δώσει άλλη τιμή στο `last_name`.

````python
add_surname("William", "Smith")
````

## Αυθαίρετος αριθμός ορισμάτων (arbitrary arguments)

Σε σπάνιες περιπτώσεις δεν θα γνωρίζετε πόσα ακριβώς ορίσματα θα χρειαστεί να δώσει ένας χρήστης στη συνάρτησή σας, οπότε θέλετε να του επιτρέψετε να δώσει όσα επιθυμεί. Τότε χρησιμοποιείτε τα λεγόμενα αυθαίρετα ορίσματα (arbitrary arguments), που ορίζονται με έναν αστερίσκο (`*`) πριν από το όνομα της παραμέτρου.

````python
def print_names(*names):
    for name in names:
        print (name)
````

````python
print_names("William", "Marge", "Sally", "Alex")
````

Παρατηρήστε ότι η συνάρτηση επέτρεψε στον χρήστη να δώσει όσα ορίσματα ήθελε. Στην πορεία μου ως προγραμματιστής έχω χρησιμοποιήσει αυθαίρετα ορίσματα μόνο λίγες φορές, αλλά είναι σημαντικό να τα έχετε υπόψη σας, για την περίπτωση που θα τα χρειαστείτε.
