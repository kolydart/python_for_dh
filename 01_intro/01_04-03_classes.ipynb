# Κλάσεις

## Εισαγωγή

Σε αυτή την ενότητα θα γνωρίσουμε τις κλάσεις. Οι **κλάσεις** (classes) μοιάζουν κάπως με τις δομές δεδομένων, αλλά διαφέρουν σε ένα σημαντικό σημείο: μπορούν να περιέχουν συναρτήσεις. Όταν δημιουργείτε μια κλάση στην Python, ουσιαστικά δημιουργείτε ένα ειδικό είδος αντικειμένου δεδομένων που μπορεί να διαθέτει συναρτήσεις. Οι συναρτήσεις που ορίζονται μέσα σε κλάσεις ονομάζονται **μέθοδοι** (methods). Στην πραγματικότητα, έχετε ήδη συναντήσει μεθόδους. Ίσως θυμάστε από το κεφάλαιο 1.2, όταν γνωρίσαμε τις συμβολοσειρές, ότι για να τροποποιήσουμε τα δεδομένα γράφαμε μια τελεία (`.`) μετά το όνομα του αντικειμένου συμβολοσειράς και στη συνέχεια τη συνάρτηση (μέθοδο) που θέλαμε να καλέσουμε, π.χ. `str1.replace(something, something_else)`. Αυτό ακριβώς διακρίνει συντακτικά μια συνάρτηση από μια μέθοδο στην Python: η συνάρτηση καλείται αυτοτελώς, ενώ η μέθοδος καλείται μέσω ενός αντικειμένου κλάσης. Η διάκριση αυτή δεν εξηγείται εύκολα, αλλά ελπίζω ότι θα γίνει σαφέστερη όταν δουλέψουμε από πιο κοντά με κλάσεις και μεθόδους παρακάτω.

## Δημιουργία κλάσης

Ας δημιουργήσουμε μια βασική κλάση. Η κλάση μας θα αποθηκεύει δεδομένα για αυτοκράτορες, οπότε ας της δώσουμε το όνομα `Emperor`. Κάθε αυτοκράτορας στο σύνολο δεδομένων μας θα έχει τέσσερα χαρακτηριστικά (attributes): όνομα, έτος γέννησης, έτος στέψης και έτος θανάτου.

````python
class Emperor:
    def __init__(self, name, birth, coronation, death):
        self.name = name
        self.birth = birth
        self.coronation = coronation
        self.death = death
````

Έτσι μοιάζει αυτή η βασική κλάση. Την πρώτη φορά ίσως δυσκολευτείτε να καταλάβετε τι συμβαίνει εδώ, οπότε ας την αναλύσουμε βήμα βήμα.

Στην πρώτη γραμμή, δηλώνουμε:

```
class Emperor:
```
Η πρώτη λέξη είναι η λέξη-κλειδί `class`. Όταν τη συναντά η Python, περιμένει να ακολουθήσει η σύνταξη ενός αντικειμένου κλάσης, όπως ακριβώς το `def` ξεκινά τον ορισμό μιας νέας συνάρτησης.

Ακολουθεί το όνομα της κλάσης, εδώ `Emperor`, και στη συνέχεια η `:`, που ολοκληρώνει την πρώτη γραμμή (μετά το όνομα μπορούν να μεσολαβούν και κενές παρενθέσεις, `Emperor():`, όπως θα δούμε παρακάτω). Η `:` σημαίνει επίσης ότι η Python περιμένει η επόμενη γραμμή να έχει εσοχή.

Η επόμενη γραμμή είναι:
```
def __init__(self, name, birth, coronation, death):
```

Αυτή είναι η δομή που θα χρησιμοποιείτε στις περισσότερες κλάσεις. Η `__init__()` είναι μια ειδική μέθοδος που εκτελείται όταν δημιουργείται ένα στιγμιότυπο (instance) της κλάσης. Μέσα στις `()` βλέπουμε τις παραμέτρους που δέχεται η συνάρτηση. Υπάρχουν πέντε παράμετροι. Η πρώτη είναι το `self`, που αναφέρεται στο ίδιο το στιγμιότυπο της κλάσης. Το `self` πρέπει να γράφεται πάντα πρώτο. Οι επόμενες τέσσερις παράμετροι είναι οι `name`, `birth`, `coronation` και `death`.

Η παράμετρος `self` επιτρέπει επίσης σε καθένα από τα χαρακτηριστικά `name`, `birth`, `coronation` και `death` να συνδέεται με ένα συγκεκριμένο στιγμιότυπο της κλάσης. Αυτό θα γίνει σαφέστερο στη συνέχεια.

Οι επόμενες τέσσερις γραμμές έχουν εσοχή και είναι οι εξής:

```
self.name = name
self.birth = birth
self.coronation = coronation
self.death = death
```

Με αυτές τις γραμμές η συνάρτηση αντιστοιχίζει τα χαρακτηριστικά στο στιγμιότυπο της κλάσης, με εντολές όπως `self.name = name` κ.ο.κ. Η κλάση μας είναι πλέον έτοιμη. Ας δημιουργήσουμε τώρα ένα αντικείμενο αυτής της κλάσης.

Ως ερευνητής της καρολίγγειας περιόδου, μελέτησα τον Καρλομάγνο, έναν από τους γνωστότερους μεσαιωνικούς βασιλείς και δεύτερο βασιλιά της δυναστείας των Καρολιγγίων. Γεννήθηκε το 742, στέφθηκε Ρωμαίος αυτοκράτορας το 800 και πέθανε το 814. Αυτά είναι τα τέσσερα βασικά δεδομένα που χρειαζόμαστε για την κλάση μας.

Μπορούμε να δημιουργήσουμε το αντικείμενο της κλάσης `Emperor` για τον Καρλομάγνο με την ακόλουθη γραμμή κώδικα.

````python
charlemagne = Emperor("Charlemagne", 742, 800, 814)
````

Παρατηρήστε ότι δημιουργούμε το αντικείμενο της κλάσης με το όνομα της κλάσης, `Emperor()`, δίνοντας τέσσερα ορίσματα που αντιστοιχούν στα τέσσερα χαρακτηριστικά: όνομα, γέννηση, στέψη, θάνατος.

Ας εκτυπώσουμε τώρα αυτό το αντικείμενο.

````python
print (charlemagne)
````

Αυτό είναι κάτι καινούριο και μάλλον όχι αυτό που περιμένατε. Δείχνει ότι πρόκειται για ειδικό αντικείμενο κλάσης. Μπορούμε να ανακτήσουμε τα δεδομένα του με τη συνάρτηση `vars()`.

````python
print(vars(charlemagne))
````

Για να προσπελάσουμε ένα συγκεκριμένο δεδομένο της κλάσης, γράφουμε το όνομα του αντικειμένου και στη συνέχεια `.name_of_attribute`. Για παράδειγμα, για το έτος γέννησης του Καρλομάγνου χρησιμοποιούμε την ακόλουθη εντολή.

````python
print(charlemagne.birth)
````

Σε αυτό το σημείο ίσως σκέφτεστε: «...και τι μ' αυτό; Απλώς έφτιαξα ένα λεξικό».

Και, προς το παρόν, έχετε δίκιο. Η κλάση μας, όπως είναι δομημένη, δεν είναι τίποτα περισσότερο από ένα λεξικό δεδομένων αποθηκευμένο με ειδικό τρόπο, και αυτό δεν είναι καλή περίπτωση χρήσης μιας κλάσης. Θυμηθείτε ότι αυτό που κάνει τις κλάσεις ξεχωριστές είναι η δυνατότητα να τους προσαρτούμε συναρτήσεις. Ας δούμε τώρα πώς γίνεται αυτό!

## Προσθήκη συναρτήσεων σε μια κλάση

Για να προσθέσουμε μια συνάρτηση σε μια κλάση, την ορίζουμε (με `def`) μέσα στο σώμα της κλάσης. Ας φτιάξουμε μια απλή συνάρτηση που εκτυπώνει μια συμβολοσειρά με το όνομα του αυτοκράτορα και το έτος γέννησής του. Για τον σκοπό αυτό, τροποποιούμε την κλάση ως εξής:

````python
class Emperor():
    def __init__(self, name, birth, coron, death):
        self.name = name
        self.birth = birth
        self.coron = coron
        self.death = death
    def birth_date(self):
        print (f"{self.name} was born in {self.birth}")
````

Παρατηρήστε ότι προσθέσαμε στην κλάση μια συνάρτηση με το όνομα `birth_date()`. Η συνάρτηση δέχεται το `self`, δηλαδή όλα τα δεδομένα του συγκεκριμένου στιγμιοτύπου, και απλώς εκτυπώνει μια f-string που δηλώνει πότε γεννήθηκε ο αυτοκράτορας.

Αρκεί τώρα να δημιουργήσουμε ένα νέο στιγμιότυπο της κλάσης για τον Καρλομάγνο.

````python
charlemagne = Emperor("Charlemagne", 742, 800, 814)
````

Μέσω αυτού του στιγμιοτύπου, καλούμε τη μέθοδο `birth_date()` ως εξής:

````python
charlemagne.birth_date()
````
