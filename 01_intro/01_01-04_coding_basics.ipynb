## Η συνάρτηση print

Το πρώτο πράγμα που μαθαίνει κάθε προγραμματιστής σε μια νέα γλώσσα είναι πώς να εκτυπώσει κάτι. Στα περισσότερα μαθήματα χρησιμοποιείται η φράση "Hello, World!". Εδώ ας δοκιμάσουμε κάτι λίγο διαφορετικό: ας υποθέσουμε ότι θέλω να εκτυπώσω "Hello, William!". Στην Python αυτό γίνεται με τη **συνάρτηση print**, η οποία εκτυπώνει ένα δεδομένο. Στην περίπτωσή μας, το δεδομένο είναι ένα κείμενο, γνωστό ως συμβολοσειρά (string, βλ. παρακάτω): "Hello, William!". Ό,τι θέλουμε να εκτυπώσουμε τοποθετείται ανάμεσα σε μια αριστερή και μια δεξιά παρένθεση. Ας εκτελέσουμε την εντολή print στο παρακάτω κελί.

```python
print("Hello, William!")
```

Για να το πετύχουμε, πρέπει να αποθηκεύσουμε ένα δεδομένο στη μνήμη. Στην Python αυτό γίνεται με τη δημιουργία μιας μεταβλητής που παραπέμπει σε ένα αντικείμενο. ## Αντικείμενα

Όταν εισάγουμε ή δημιουργούμε δεδομένα στην Python, ουσιαστικά δημιουργούμε ένα αντικείμενο στη μνήμη και μια μεταβλητή που παραπέμπει σε αυτό. Οι δύο αυτές λέξεις, <b>αντικείμενο</b> (object) και <b>μεταβλητή</b> (variable), σημαίνουν ελαφρώς διαφορετικά πράγματα, αλλά συχνά χρησιμοποιούνται εναλλακτικά. Το εγχειρίδιο δεν θα υπεισέλθει στις λεπτές διαφορές τους και στους λόγους ύπαρξής τους. Προς το παρόν, φανταστείτε το αντικείμενο ως κάτι που δημιουργείται από ένα σενάριο (script) Python και αποθηκεύεται στη μνήμη του υπολογιστή, ώστε να μπορεί να χρησιμοποιηθεί αργότερα σε ένα πρόγραμμα.

Σκεφτείτε τη <b>μνήμη του υπολογιστή σας</b> σαν τον εγκέφαλό σας. Ας υποθέσουμε ότι πρέπει να θυμάστε πώς λέγεται το «hello» στα γερμανικά. Μπορεί να χρησιμοποιήσετε τη μνήμη σας όπως μια κάρτα εκμάθησης λεξιλογίου (flashcard), όπου το αγγλικό «hello» αντιστοιχεί στο γερμανικό «hallo». Στην Python δημιουργούμε αντικείμενα με παρόμοιο τρόπο. Το αντικείμενο εδώ θα ήταν η λημματική καταχώριση «hello: hallo».

## Μεταβλητές

Για να αναφερθούμε σε αυτή την καταχώριση της μνήμης, χρειαζόμαστε έναν τρόπο να την καλούμε. Αυτό γίνεται με μια μεταβλητή. Η μεταβλητή είναι απλώς ένα όνομα μέσα στο σενάριό μας, το οποίο παραπέμπει στο αντικείμενο που βρίσκεται στη μνήμη. Χάρη στις μεταβλητές, έχουμε ένα εύκολο στην απομνημόνευση όνομα για να αναφερόμαστε σε αυτό το αντικείμενο, να το καλούμε και να το τροποποιούμε.

Μια μεταβλητή δημιουργείται αν πληκτρολογήσουμε ένα μοναδικό όνομα, ακολουθούμενο από το σύμβολο = και, στη συνέχεια, από τα δεδομένα. Όπως θα δούμε σε αυτό το κεφάλαιο, υπάρχουν πολλοί τύποι δεδομένων, που ο καθένας δημιουργείται με διαφορετικό τρόπο. Ας δημιουργήσουμε το πρώτο μας αντικείμενο: μια συμβολοσειρά, δηλαδή ένα κομμάτι κειμένου (θα τις εξετάσουμε αναλυτικότερα παρακάτω). Θα δημιουργήσουμε το αντικείμενο `synthetis` και θα του αντιστοιχίσουμε στη μνήμη το όνομα ενός συνθέτη. Ας το κάνουμε στο παρακάτω κελί (μπλοκ κώδικα).

```python
synthetis = "Μανώλης Καλομοίρης"
```

Εξαιρετικά! Δημιουργήσαμε το πρώτο μας αντικείμενο. Ήρθε η ώρα να το χρησιμοποιήσουμε. Παρακάτω θα μάθουμε διάφορους τρόπους χειρισμού των συμβολοσειρών· προς το παρόν, ας ελέγξουμε απλώς αν το αντικείμενο υπάρχει στη μνήμη. Αυτό γίνεται με τη συνάρτηση print.

Η συνάρτηση print θα γίνει ο καλύτερος φίλος σας στην Python. Είναι ίσως η συνάρτηση που χρησιμοποιώ συχνότερα, επειδή διευκολύνει την αποσφαλμάτωση (debugging), δηλαδή τον εντοπισμό και τη διόρθωση προβλημάτων στον κώδικα. Μας επιτρέπει να εκτυπώνουμε αντικείμενα που είναι αποθηκευμένα στη μνήμη.

Για να χρησιμοποιήσουμε τη συνάρτηση print, πληκτρολογούμε τη λέξη print ακολουθούμενη από μια αριστερή παρένθεση. Μετά την παρένθεση τοποθετούμε το αντικείμενο ή το δεδομένο που θέλουμε να εκτυπώσουμε και, τέλος, κλείνουμε τη συνάρτηση με μια δεξιά παρένθεση. Ας εκτυπώσουμε το νέο μας αντικείμενο `synthetis`, για να βεβαιωθούμε ότι βρίσκεται στη μνήμη.

```python
print(synthetis)
```

Παρατηρήστε ότι, όταν εκτελώ το παραπάνω κελί, βλέπω ως έξοδο το περιεχόμενο του αντικειμένου που δημιουργήσαμε. Τι θα συμβεί αν προσπαθήσω να εκτυπώσω το ίδιο αντικείμενο με κεφαλαίο το πρώτο γράμμα, δηλαδή `Synthetis` αντί για `synthetis`;

## Διάκριση πεζών και κεφαλαίων

```python
print(Synthetis)
```

Το τρομακτικό μπλοκ κειμένου παραπάνω δείχνει ότι προκαλέσαμε ένα σφάλμα στην Python. Το σφάλμα αυτό μας διδάσκει δύο πράγματα. Πρώτον, η Python κάνει διάκριση πεζών και κεφαλαίων (case sensitive). Επομένως, για να ταιριάζει ένα όνομα αντικειμένου (ή μια συμβολοσειρά), πρέπει να συμπίπτουν όχι μόνο τα γράμματα αλλά και το αν είναι πεζά ή κεφαλαία. Δεύτερον, μπορούμε να καλέσουμε μόνο αντικείμενα που έχουν δημιουργηθεί και αποθηκευτεί στη μνήμη.

Τώρα που η μεταβλητή μας παραπέμπει σε ένα συγκεκριμένο δεδομένο, μπορούμε να κάνουμε την παραπάνω εντολή print λίγο πιο δυναμική. Ας εκτυπώσουμε μια φράση που περιέχει το όνομα του συνθέτη από τη μνήμη. Δεν περιμένω να κατανοήσετε τις λεπτομέρειες του παρακάτω κώδικα· αρκεί να κρατήσετε ότι συχνά θα χρειαστεί να αποθηκεύουμε μεταβλητές στη μνήμη, ώστε να τις χρησιμοποιούμε αργότερα στα προγράμματά μας.

```python
print(f"Αγαπημένος μου συνθέτης: {synthetis}")
```

## Δεσμευμένες λέξεις

Η Python διαθέτει ορισμένες λέξεις γνωστές ως **δεσμευμένες λέξεις** (reserved words ή keywords), οι οποίες δεν μπορούν να χρησιμοποιηθούν ως ονόματα μεταβλητών. Στην έκδοση 3.6 της Python υπάρχουν συνολικά 33 δεσμευμένες λέξεις. Επειδή είναι μερικές φορές δύσκολο να τις θυμάστε όλες, η Python διαθέτει μια χρήσιμη ενσωματωμένη συνάρτηση, την `help`. Αν εκτελέσουμε την ακόλουθη εντολή, θα εμφανιστεί ολόκληρος ο κατάλογος.

```python
help("keywords")
```

Αυτές είναι οι λέξεις που δεν μπορείτε να χρησιμοποιήσετε ως ονόματα μεταβλητών.

## Ενσωματωμένοι τύποι

Εκτός από τις δεσμευμένες λέξεις, η Python διαθέτει και ενσωματωμένους τύπους (built-in types). Πρόκειται για λέξεις που μπορείτε να χρησιμοποιήσετε (όπως θα δούμε) για να μετατρέψετε έναν τύπο δεδομένων σε άλλον. Υπάρχουν συνολικά 94. Σε αντίθεση με τις δεσμευμένες λέξεις, *μπορείτε* να χρησιμοποιήσετε έναν ενσωματωμένο τύπο ως όνομα μεταβλητής. Ωστόσο, αυτό αποθαρρύνεται έντονα, γιατί έτσι αντικαθίσταται η αρχική λειτουργία του ονόματος μέσα στο σενάριό σας.

```python
import builtins
[getattr(builtins, d) for d in dir(builtins) if isinstance(getattr(builtins, d), type)]
```

Από αυτή τη μεγάλη λίστα, συνιστώ να προσέξετε ιδιαίτερα όσα ονόματα είναι πιθανότερο να γράψετε αυθόρμητα: bool, dict, float, int, list, map, object, property, range, reversed, set, slice, str, super, tuple, type και zip. Είναι πολύ πιθανότερο να χρησιμοποιήσετε κατά λάθος κάποιο από αυτά ως όνομα μεταβλητής απ' ό,τι, ας πούμε, το ZeroDivisionError· και μια μικρότερη λίστα απομνημονεύεται πολύ ευκολότερα.

## Η συνάρτηση type

Συχνά χρειάζεται να ελέγξουμε σε ποιον τύπο δεδομένων ανήκει μια μεταβλητή. Για τον σκοπό αυτό χρησιμοποιείται η ενσωματωμένη συνάρτηση **type** της Python. Γράφουμε την εντολή όπως παρακάτω, τοποθετώντας μέσα στις παρενθέσεις τα δεδομένα που θέλουμε να εξετάσουμε.

```python
type("this is a string...")
```

## Σφάλματα (bugs)

Σε όλη την προγραμματιστική σας πορεία θα διαβάζετε ή θα ακούτε συχνά για `bugs` (σφάλματα). Ένα bug είναι ένα πρόβλημα στον κώδικα που προκαλεί είτε μήνυμα σφάλματος είτε απρόσμενο αποτέλεσμα στην έξοδο. Ο εντοπισμός και η διόρθωση των bugs ονομάζεται `debugging` (αποσφαλμάτωση). Πέρα από τη διατύπωση λύσεων σε κώδικα, το `debugging` μπορεί να είναι από τα πιο χρονοβόρα μέρη της συγγραφής ενός προγράμματος, ιδίως όταν αυτό είναι αρκετά σύνθετο. Σε όλο το εγχειρίδιο θα συναντήσουμε συνηθισμένα σφάλματα, ώστε να τα δείτε σε ένα ελεγχόμενο περιβάλλον, και θα εξηγήσουμε τι σημαίνει το καθένα και πώς διορθώνεται. Ωστόσο, είναι πιθανό να προκαλέσετε πολλά άλλα bugs όταν προσπαθήσετε να εφαρμόσετε τον κώδικα του εγχειριδίου στα δικά σας δεδομένα. Αυτό είναι αναμενόμενο. Να διαβάζετε πάντα προσεκτικά κάθε γραμμή του κώδικά σας και, όταν εμφανίζεται ένα σφάλμα, να εντοπίζετε από πού προέρχεται και τι είδους σφάλμα είναι.
