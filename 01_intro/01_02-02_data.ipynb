# Εισαγωγή στα δεδομένα

## Τι είναι τα δεδομένα;

Στην Python θα δουλέψουμε με επτά βασικούς τύπους δεδομένων και δομών δεδομένων:

- συμβολοσειρές (κείμενο)
- ακέραιοι (ακέραιοι αριθμοί)
- δεκαδικοί (αριθμοί κινητής υποδιαστολής, float)
- Boolean (λογικές τιμές: True ή False)
- λίστες
- πλειάδες (λίστες που δεν μπορούν να τροποποιηθούν στη μνήμη)
- λεξικά (κλειδί : τιμή)

Στο κεφάλαιο αυτό θα εξετάσουμε καθένα από αυτά. Η παρούσα ενότητα εστιάζει στα δεδομένα (συμβολοσειρές, ακέραιοι, δεκαδικοί και Boolean), ενώ η επόμενη στις δομές δεδομένων (λίστες, πλειάδες και λεξικά).

Τα δεδομένα (data, στον ενικό datum) είναι μονάδες πληροφορίας, όπως οι ακέραιοι, οι δεκαδικοί και οι συμβολοσειρές. Οι δομές δεδομένων είναι αντικείμενα που οργανώνουν τα δεδομένα και τα συσχετίζουν μεταξύ τους, όπως οι λίστες, οι πλειάδες και τα λεξικά. Αρχίστε να εξασκείτε το μάτι σας ώστε να αναγνωρίζει τη σύνταξη της Python για τα δεδομένα και τις δομές δεδομένων που εξετάζονται παρακάτω.

Για ευκολία, ο παρακάτω πίνακας συνοψίζει αυτούς τους τύπους δεδομένων, ώστε να ανατρέχετε σε αυτόν όποτε χρειάζεται.

`````{admonition} Τύποι δεδομένων και δομές δεδομένων στην Python
:class: tip
| Όνομα        | Τύπος   | Κατηγορία  | Παράδειγμα        | Πώς αναγνωρίζεται       | Μεταβαλλόμενο |
|--------------|---------|------------|-------------------|-------------------------|---------------|
| συμβολοσειρά | `str`   | κείμενο    | `"William"`       | εισαγωγικά              | όχι           |
| ακέραιος     | `int`   | αριθμός    | `1`               | χωρίς δεκαδικό μέρος    | όχι           |
| δεκαδικός    | `float` | αριθμός    | `1.1`             | με δεκαδικό μέρος       | όχι           |
| Boolean      | `bool`  | λογική τιμή | `True`           | `True` ή `False`        | όχι           |
| λίστα        | `list`  | ακολουθία  | `[1, 1.1, "one"]` | `[]`                    | ναι           |
| πλειάδα      | `tuple` | ακολουθία  | `(1, 1.1, "one")` | `()`                    | όχι           |
`````

## Εργασία με συμβολοσειρές ως δεδομένα

Όταν εργάζεστε με την Python, συχνά δεν θα δημιουργείτε απλώς δεδομένα, αλλά θα τα χειρίζεστε και θα τα τροποποιείτε. Επειδή οι ερευνητές των ανθρωπιστικών επιστημών εργάζονται συχνά με συμβολοσειρές, σας συνιστώ να αφιερώσετε αρκετό χρόνο στην εξάσκηση και την εξοικείωση με τους βασικούς τρόπους χειρισμού τους στην Python μέσω των ενσωματωμένων μεθόδων.

Για να χειριστούμε τις συμβολοσειρές ως δεδομένα, χρησιμοποιούμε <b>μεθόδους</b> (methods) και <b>συναρτήσεις</b> (functions). Οι βασικές συναρτήσεις για τον χειρισμό συμβολοσειρών είναι ενσωματωμένες στην Python, οπότε δεν χρειάζεται να εγκαταστήσετε βιβλιοθήκες τρίτων. Αργότερα στο εγχειρίδιο θα κάνουμε πιο προχωρημένες εργασίες με συμβολοσειρές χρησιμοποιώντας βιβλιοθήκες τρίτων, όπως η regex· προς το παρόν, θα περιοριστούμε στις βασικές μεθόδους.

Ας δούμε τώρα πώς χειριζόμαστε τις συμβολοσειρές μέσω κώδικα. Πρώτα πρέπει να δημιουργήσουμε μια συμβολοσειρά· ας αποθηκεύσουμε σε αυτήν το όνομα ενός συνθέτη.

``````python
synthetis = "Δημήτρης Μητρόπουλος"
``````

### Η μέθοδος upper

Ας μετατρέψουμε τώρα ολόκληρη τη συμβολοσειρά σε κεφαλαία. Αυτό γίνεται με τη μέθοδο <b>`.upper()`</b>. Παρατηρήστε ότι το `.upper()` γράφεται μετά τη συμβολοσειρά και ότι μέσα στις `()` δεν υπάρχουν ορίσματα. Έτσι μπορείτε εύκολα να αναγνωρίσετε μια μέθοδο (σε αντίθεση με μια συνάρτηση). Θα μάθουμε περισσότερα για αυτή τη διάκριση στα κεφάλαια για τις συναρτήσεις και τις κλάσεις.

``````python
print(synthetis.upper())
``````

### Η μέθοδος lower

Παρατηρήστε ότι η συμβολοσειρά μας είναι τώρα όλη με κεφαλαία. Κοιτάξτε όμως προσεκτικά: η Python κράτησε τους τόνους, ενώ η ελληνική ορθογραφία τους παραλείπει στα κεφαλαία. Οι μέθοδοι αυτές δουλεύουν «σχεδόν» σωστά στα ελληνικά, και το «σχεδόν» το εντοπίζει μόνο όποιος ξέρει τη γλώσσα.

Το αντίστροφο κάνει η μέθοδος <b>`.lower()`</b>: μετατρέπει όλη τη συμβολοσειρά σε πεζά. Ας τη δοκιμάσουμε σε ένα όνομα γραμμένο όλο με κεφαλαία, όπως εμφανίζεται συχνά σε παλαιότερους καταλόγους.

``````python
print("ΜΗΤΡΟΠΟΥΛΟΣ ΔΗΜΗΤΡΗΣ".lower())
``````

Το τελικό σίγμα (`ς`) είναι σωστό, οι τόνοι όμως λείπουν. Η πληροφορία δεν υπάρχει στα δεδομένα, και η Python δεν μπορεί να την προσθέσει.

### Η μέθοδος capitalize

Εκ πρώτης όψεως, αυτές οι μέθοδοι μοιάζουν χρήσιμες μόνο σε ειδικές περιπτώσεις. Πέρα όμως από το να δίνουν στις συμβολοσειρές την εμφάνιση που θέλετε, έχουν πολύ ευρύτερη χρησιμότητα. Ας υποθέσουμε ότι θέλετε να αναζητήσετε ένα όνομα, "Μητρόπουλος", σε μια συμβολοσειρά. Τι γίνεται αν τα δεδομένα σας προέρχονται από μηνύματα ηλεκτρονικού ταχυδρομείου, γραπτά μηνύματα κ.λπ.; Το "Μητρόπουλος" μπορεί να είναι γραμμένο με κεφαλαίο ή με πεζό αρχικό γράμμα, οπότε θα έπρεπε να εκτελέσετε δύο αναζητήσεις. Αν όμως μετατρέψετε τη συμβολοσειρά σε πεζά πριν από την αναζήτηση, αρκεί να αναζητήσετε "μητρόπουλος" και θα βρείτε όλες τις εμφανίσεις. Αυτό ακριβώς κάνουν στο παρασκήνιο (back-end) οι περισσότερες μηχανές αναζήτησης, ώστε η αναζήτηση να μην κάνει διάκριση πεζών και κεφαλαίων. Στην Python, ωστόσο, το βήμα αυτό του καθαρισμού δεδομένων πρέπει να το εκτελέσετε εσείς πριν από κάθε αναζήτηση σε συμβολοσειρές.

Ας δούμε μια άλλη μέθοδο, την <b>`.capitalize()`</b>, η οποία μετατρέπει το πρώτο γράμμα μιας συμβολοσειράς σε κεφαλαίο.

``````python
epitheto = "ΜΗΤΡΟΠΟΥΛΟΣ"
``````

``````python
print(epitheto.capitalize())
``````

Η `.capitalize()` κάνει κεφαλαίο μόνο το πρώτο γράμμα της συμβολοσειράς και όλα τα υπόλοιπα πεζά. Μια συγγενική μέθοδος, η <b>`.title()`</b>, κάνει κεφαλαίο το πρώτο γράμμα κάθε λέξης:

``````python
print("ΜΗΤΡΟΠΟΥΛΟΣ ΔΗΜΗΤΡΗΣ".title())
``````

Τέτοιες μέθοδοι χρησιμεύουν στον καθαρισμό δεδομένων, όταν τα κύρια ονόματα ενός συνόλου δεδομένων πρέπει να έχουν όλα την ίδια μορφή. Στα ελληνικά, όμως, το αποτέλεσμα θέλει πάντα έλεγχο: και οι δύο μέθοδοι έδωσαν `Μητροπουλος`, χωρίς τόνο.

### Η μέθοδος replace

Ίσως η πιο χρήσιμη μέθοδος συμβολοσειρών είναι η <b>`.replace()`</b>. Γιατί είναι τόσο χρήσιμη; Αν χρησιμοποιείτε την Python για την ανάλυση κειμένων, σας εγγυώμαι ότι τα κείμενα δεν θα είναι ποτέ πραγματικά καθαρά. Μπορεί να έχουν λανθασμένη κωδικοποίηση, χαρακτήρες που αλλοιώνουν τα αποτελέσματα των αναζητήσεων, κακής ποιότητας OCR, πολλαπλές αλλαγές γραμμής, λέξεις που έχουν κοπεί με ενωτικό, και ο κατάλογος συνεχίζεται. Η μέθοδος `.replace()` σας επιτρέπει να καθαρίσετε γρήγορα και αποτελεσματικά τα κειμενικά δεδομένα, ώστε να αποκτήσουν ενιαία μορφή.

Σε αντίθεση με τις προηγούμενες μεθόδους, η `.replace()` δέχεται υποχρεωτικά δύο ορίσματα (arguments), δηλαδή τιμές που γράφονται μέσα στις παρενθέσεις, χωρισμένες με κόμμα. Η σειρά τους έχει σημασία: το πρώτο όρισμα είναι η υποσυμβολοσειρά (το τμήμα της συμβολοσειράς) που θέλετε να αντικαταστήσετε και το δεύτερο είναι αυτό με το οποίο θα αντικατασταθεί. Και τα δύο πρέπει να είναι συμβολοσειρές. Η γενική μορφή είναι η εξής:

```
.replace("τμήμα προς αντικατάσταση", "νέο τμήμα")
```

Στο παρακάτω παράδειγμα, ας αντικαταστήσουμε την τελεία στο τέλος του "Mattingly."

``````python
introduction = "My name is William Mattingly."
``````

``````python
print(introduction.replace(".", ""))
``````

Εξαιρετικά! Ας εκτυπώσουμε τώρα ξανά τη μεταβλητή `introduction` για να δούμε τι συμβαίνει.

``````python
print(introduction)
``````

Ωχ! Κάτι δεν πάει καλά εδώ... Δεν άλλαξε τίποτα! Αυτό συμβαίνει επειδή οι συμβολοσειρές είναι <b>αμετάβλητα αντικείμενα</b> (immutable), δηλαδή αντικείμενα που δεν μπορούν να τροποποιηθούν στη μνήμη. Όπως θα δούμε στο επόμενο κεφάλαιο με τις λίστες, υπάρχουν και αντικείμενα που <i>μπορούν</i> να τροποποιηθούν στη μνήμη· αυτά ονομάζονται <b>μεταβαλλόμενα αντικείμενα</b> (mutable). Για να αλλάξετε μια συμβολοσειρά, επομένως, πρέπει να τη δημιουργήσετε ξανά στη μνήμη ή να δημιουργήσετε από αυτήν ένα νέο αντικείμενο συμβολοσειράς. Ας το δοκιμάσουμε παρακάτω.

``````python
new_introduction = introduction.replace(".", "")
``````

``````python
print(new_introduction)
``````

### Η μέθοδος split

Οι συμβολοσειρές διαθέτουν πολλές άλλες χρήσιμες μεθόδους, τις οποίες θα γνωρίσουμε σε όλο το εγχειρίδιο. Μία από αυτές είναι η `split()`, η οποία επιστρέφει μια λίστα υποσυμβολοσειρών (μικρότερων συμβολοσειρών), χωρισμένων στα σημεία όπου εμφανίζεται το διαχωριστικό (separator), δηλαδή το όρισμα της μεθόδου. Το διαχωριστικό δείχνει στην Python πού να χωρίσει τη συμβολοσειρά. Από προεπιλογή, η `split()` χωρίζει τη συμβολοσειρά στα κενά. Στους καταλόγους των βιβλιοθηκών τα ονόματα γράφονται «Επώνυμο, Όνομα», για να ταξινομούνται κατά επώνυμο. Ας προσπαθήσουμε να χωρίσουμε ένα τέτοιο όνομα.

``````python
onoma = "Σκαλκώτας, Νίκος"
``````

Όπως η `replace()`, έτσι και η `split()` είναι μέθοδος· επομένως τη γράφουμε μετά το όνομα της μεταβλητής, χωρισμένη με τελεία.

``````python
print(onoma.split())
``````

Το όνομα χωρίστηκε σε μεμονωμένες λέξεις χάρη στο προεπιλεγμένο διαχωριστικό της μεθόδου `split()`, δηλαδή το κενό. Παρατηρήστε ότι το διαχωριστικό δεν εμφανίζεται στη λίστα των υποσυμβολοσειρών, όμως το κόμμα έμεινε κολλημένο στο επώνυμο: για την Python είναι απλώς ένας ακόμη χαρακτήρας της πρώτης λέξης. Στο επόμενο κεφάλαιο θα γνωρίσουμε τις δομές δεδομένων, όπως οι λίστες, και θα μάθουμε πώς ανακτούμε συγκεκριμένα στοιχεία μιας λίστας μέσω της θέσης τους (δείκτη, index).

Η `split()` μπορεί επίσης να δεχτεί ένα όρισμα που ορίζει πού θα χωριστεί η συμβολοσειρά, για παράδειγμα σε κάποιον χαρακτήρα διαφορετικό από το κενό. Εδώ το κατάλληλο διαχωριστικό είναι το κόμμα μαζί με το κενό που ακολουθεί, `", "`.

``````python
print(onoma.split(", "))
``````

Όπως βλέπουμε, το επώνυμο χωρίστηκε με επιτυχία από το όνομα. Στο κεφάλαιο για τους βρόχους θα τα ενώσουμε ξανά με την αντίστροφη σειρά, «Όνομα Επώνυμο». Θα δουλέψουμε πολλές ακόμη φορές με συμβολοσειρές στη συνέχεια του εγχειριδίου. Σε αυτό το σημείο θα πρέπει να γνωρίζετε τι είναι οι συμβολοσειρές, πώς δημιουργούνται και πώς τις χειριζόμαστε με ορισμένες βασικές μεθόδους, όπως οι `upper()`, `lower()`, `capitalize()`, `title()`, `replace()` και `split()`.

## Αριθμοί (ακέραιοι και δεκαδικοί)

Οι γλώσσες προγραμματισμού αναπαριστούν τους αριθμούς με διάφορους τρόπους. Εδώ θα εξετάσουμε δύο: τους ακέραιους και τους δεκαδικούς.

Ένας <b>ακέραιος</b> (integer) είναι ένας αριθμός χωρίς δεκαδικό μέρος, π.χ. 1, 2 ή 3, και μπορεί να έχει οποιοδήποτε μέγεθος, π.χ. `100001200`. Ένας <b>δεκαδικός</b> (float) είναι, αντίθετα, ένας αριθμός με δεκαδικό μέρος. Έτσι, το `1` είναι ακέραιος, ενώ το `1.0` είναι δεκαδικός. Στην Python, όπως στα αγγλικά, το δεκαδικό μέρος χωρίζεται με τελεία και όχι με κόμμα. Οι δεκαδικοί, όπως και οι ακέραιοι, μπορούν να έχουν οποιοδήποτε μέγεθος, αλλά έχουν πάντα δεκαδικό μέρος, π.χ. `200.0020002938`. Για να δημιουργήσετε ένα αντικείμενο ακέραιου ή δεκαδικού αριθμού δεν χρειάζεστε ειδικούς χαρακτήρες· αρκεί το σύμβολο της ισότητας. Στο παρακάτω παράδειγμα δημιουργούμε δύο αντικείμενα, καθένα με ένα σύμβολο ισότητας: το `an_integer`, που αντιστοιχεί στον ακέραιο `1`, και το `a_float`, που αντιστοιχεί στον δεκαδικό `1.1`.

## Εργασία με αριθμούς ως δεδομένα

Αφού γνωρίσατε τις συμβολοσειρές, ας περάσουμε σε έναν άλλο τύπο δεδομένων: τους αριθμούς. Στην Python οι αριθμοί είναι δύο κυρίως ειδών:

## Λογικές τιμές (Boolean)

Ο όρος boolean προέρχεται από την άλγεβρα Boole (λογική άλγεβρα), έναν κλάδο των μαθηματικών που βασίζεται στη δυαδική λογική. Το δυαδικό σύστημα αποτελεί τη βάση όλων των υπολογιστών, με εξαίρεση τους νεότερους κβαντικούς υπολογιστές. Στο δυαδικό σύστημα κάθε τιμή είναι 0 ή 1· απενεργοποιημένο ή ενεργοποιημένο· ψευδές ή αληθές. Στις γλώσσες προγραμματισμού, ένα αντικείμενο boolean είναι είτε `True` (αληθές) είτε `False` (ψευδές): το `True` αντιστοιχεί στο 1 και το `False` στο 0. Στην Python οι τιμές αυτές γράφονται με κεφαλαίο το αρχικό γράμμα, T ή F, δηλαδή `True` ή `False`. Ας δημιουργήσουμε ένα τέτοιο αντικείμενο.
