## Τι είναι οι βρόχοι;

Οι βρόχοι (loops) είναι θεμελιώδης έννοια σε όλες τις γλώσσες προγραμματισμού και απαραίτητοι σχεδόν σε κάθε κομμάτι κώδικα που γράφουμε. Είναι χρήσιμοι επειδή μας επιτρέπουν να επαναλαμβάνουμε συστηματικά ορισμένα βήματα. **Επανάληψη** (iteration) ονομάζεται η διαδικασία με την οποία διατρέχουμε ένα σύνολο ή μια συλλογή δεδομένων. Όπως θα δούμε σε αυτό το κεφάλαιο, η Python διαθέτει δύο τύπους βρόχων:

1. Βρόχοι For (For Loops)
2. Βρόχοι While (While Loops)

Οι περισσότερες εργασίες μπορούν να γίνουν με οποιονδήποτε από τους δύο βρόχους, αλλά ο τρόπος χρήσης τους διαφέρει. Στο τέλος του κεφαλαίου δεν θα ξέρετε μόνο πώς δημιουργείται ένας βρόχος στην Python, αλλά θα έχετε και μια γενική εικόνα για το πότε ενδείκνυται ο καθένας. Όσο γράφετε περισσότερο κώδικα, θα διαπιστώσετε ότι οι προγραμματιστές τείνουν να προτιμούν τον έναν τύπο βρόχου από τον άλλο. Εγώ προτιμώ τον βρόχο for. Αν με ρωτούσατε γιατί, δεν θα μπορούσα να σας απαντήσω. Ίσως στο τέλος του κεφαλαίου να έχετε κι εσείς τον αγαπημένο σας!

## Βρόχοι for

Για να κατανοήσετε τον βρόχο for, φανταστείτε πρώτα μια λίστα ονομάτων από τον κατάλογο μιας βιβλιοθήκης:

- Μάντζαρος, Νικόλαος
- Σαμάρας, Σπυρίδων
- Καλομοίρης, Μανώλης
- Σκαλκώτας, Νίκος

Παρατηρήστε ότι η παραπάνω λίστα περιέχει τέσσερα ονόματα, καθένα στη μορφή «Επώνυμο, Όνομα». Ο βρόχος for μάς επιτρέπει να διατρέξουμε αυτή τη λίστα, από το ένα όνομα στο επόμενο. Ακόμη και σήμερα, όταν γράφω έναν βρόχο for, λέω μέσα μου:

"Για κάθε στοιχείο σε αυτή τη λίστα, κάνε κάτι."

Ας υποθέσουμε ότι θέλουμε να εκτυπώσουμε με την Python καθένα από τα ονόματα. Θα έλεγα μέσα μου:

"Για κάθε στοιχείο σε αυτή τη λίστα, εκτύπωσε το όνομα".

Δεν έχω γράψει ούτε μία γραμμή κώδικα, αλλά έχω διατυπώσει μια λογική, δηλαδή μια σειρά ενεργειών, σε φυσική γλώσσα. Αυτό ονομάζεται **ψευδοκώδικας** (pseudo-code). Ο ψευδοκώδικας είναι εξαιρετικός τρόπος σκέψης για τον κώδικα, επειδή είναι ανεξάρτητος από τη γλώσσα προγραμματισμού: ο ίδιος ψευδοκώδικας μπορεί να υλοποιηθεί τόσο σε Python όσο και σε C. Είναι χρήσιμος επειδή εστιάζει στη λογική των εντολών και όχι στον κώδικα. Όσο γράφετε πιο σύνθετα προγράμματα, θα διαπιστώνετε ότι η δυσκολία και τα σφάλματα δεν προέρχονται πάντα από τη γλώσσα προγραμματισμού, αλλά από τη λογική με την οποία την έχετε εφαρμόσει. Ο ψευδοκώδικας σας επιτρέπει να καταγράψετε πρώτα τη λογική αυτού που θέλετε να κάνετε και έπειτα να εστιάσετε στην υλοποίησή της.

Η παραπάνω διατύπωση σε φυσική γλώσσα («Για κάθε στοιχείο σε αυτή τη λίστα, εκτύπωσε το όνομα») δεν θα λειτουργήσει αν τη γράψω σε ένα αρχείο Python, μου επιτρέπει όμως να καταγράψω στο χαρτί τη λογική αυτού που θέλω να κάνω. Στη συνέχεια, πρέπει να εκφράσω αυτή τη λογική με την κανονική σύνταξη της Python. Ας το κάνουμε τώρα.

Πρώτα, ας δημιουργήσουμε μια λίστα ονομάτων, δηλαδή ένα αντικείμενο με το όνομα `katalogos`. Σε ονόματα μεταβλητών με περισσότερες λέξεις, οι λέξεις χωρίζονται συνήθως με κάτω παύλα (`name_list`) ή με κεφαλαίο γράμμα (`nameList`). Και οι δύο συμβάσεις είναι αποδεκτές.

````python
katalogos = ["Μάντζαρος, Νικόλαος", "Σαμάρας, Σπυρίδων", "Καλομοίρης, Μανώλης", "Σκαλκώτας, Νίκος"]
````

Εξαιρετικά! Τώρα που έχουμε τη μεταβλητή `katalogos`, ας τη διατρέξουμε, δηλαδή ας περάσουμε από κάθε στοιχείο της λίστας. Στην Python αυτό γράφεται ως εξής:

````python
for onoma in katalogos:
    print(onoma)
````

Ας αναλύσουμε τι συμβαίνει εδώ. Ξεκινάμε γράφοντας `for`. Αυτό δηλώνει στην Python ότι ξεκινά ένας βρόχος for και ότι πρέπει να ελέγξει αν η σύνταξη είναι σωστή. Ακολουθεί η λέξη `onoma`. Πρόκειται για μια μεταβλητή που παραπέμπει σε ένα αντικείμενο στη μνήμη και αλλάζει σε κάθε επανάληψη του βρόχου, καθώς ο βρόχος προχωρά στο επόμενο στοιχείο του `katalogos`. Η λέξη `katalogos` είναι το όνομα του αντικειμένου που θέλω να διατρέξω. Τέλος, υπάρχει η άνω και κάτω τελεία (`:`). Στην Python, με αυτήν δηλώνουμε ότι ακολουθεί ένα ένθετο μπλοκ κώδικα. Θα τη χρησιμοποιούμε σχεδόν σε κάθε σενάριο, επειδή συχνά χρειάζεται να εμφωλεύουμε τμήματα κώδικα. Μόλις η Python συναντήσει την `:`, περιμένει η επόμενη γραμμή να έχει εσοχή.

Η επόμενη γραμμή ξεκινά με εσοχή, ακολουθούμενη από την `print`. Όπως θυμάστε, η συνάρτηση print εκτυπώνει κάτι στην έξοδο. Ακολουθεί αυτό που θέλουμε να εκτυπωθεί, η μεταβλητή `onoma`. Με άλλα λόγια, θέλουμε να εκτυπώσουμε κάθε όνομα που αποθηκεύεται προσωρινά στη μνήμη.

Παρατηρήστε ότι στο παρακάτω μπλοκ κώδικα έχω αλλάξει το `onoma` σε `item`. Το όνομα εδώ δεν έχει σημασία· είναι απλώς η λέξη με την οποία ονομάζουμε το αντικείμενο που δημιουργείται. Θεωρείται ωστόσο Pythonic, δηλαδή καλή πρακτική, να επιλέγετε ένα όνομα με νόημα. Επειδή πρόκειται για λίστα ονομάτων, είναι λογικό να χρησιμοποιήσουμε τη λέξη `onoma` (όνομα), ώστε όσοι διαβάζουν τον κώδικά σας να τον κατανοούν ευκολότερα.

````python
for item in katalogos:
    print(item)
````

Η εκτύπωση μέσα σε έναν βρόχο for είναι χρήσιμη, ιδίως κατά την αποσφαλμάτωση, αλλά συνήθως οι βρόχοι for χρησιμοποιούνται για την τροποποίηση δεδομένων. Στο παράδειγμα παρακάτω, θέλω να δημιουργήσω μια νέα λίστα με βάση τον katalogos, με τα ίδια ονόματα στη μορφή «Όνομα Επώνυμο», όπως τα γράφουμε σε ένα κείμενο. Θα μπορούσα να ξαναγράψω με το χέρι κάθε όνομα, αλλά αυτό θα ήταν κουραστικό και, αν ο κατάλογος είχε χιλιάδες ονόματα, αδύνατο.

Μέσα στον βρόχο for, επομένως, χωρίζω κάθε όνομα στο `", "` με τη μέθοδο `split()` (κεφ. 1.2). Παίρνω έτσι μια λίστα δύο στοιχείων, την `kommatia`: το `kommatia[0]` είναι το επώνυμο και το `kommatia[1]` το όνομα. Τα ενώνω με την αντίστροφη σειρά με τον απλούστερο τρόπο, δηλαδή με το σύμβολο `+` ανάμεσα στις συμβολοσειρές και ένα κενό στη μέση, και προσθέτω το αποτέλεσμα στη νέα λίστα `nea_onomata`.

````python
nea_onomata = []
for onoma in katalogos:
    kommatia = onoma.split(", ")
    print(kommatia[1] + " " + kommatia[0])
    nea_onomata.append(kommatia[1] + " " + kommatia[0])
````

Αφού εκτελέσαμε το παραπάνω κελί, ας εκτυπώσουμε τη νέα λίστα για να δούμε τη μορφή της.

````python
print(nea_onomata)
````

Ιδού! Με έναν βρόχο for, ξαναγράψαμε όλα τα ονόματα του καταλόγου.

## List comprehension

Σε κώδικα άλλων προγραμματιστών θα δείτε μερικές φορές την ίδια ιδέα εκφρασμένη σε μία μόνο γραμμή, ως εξής:

````python
nea_onomata_comprehension = [onoma.split(", ")[1] + " " + onoma.split(", ")[0] for onoma in katalogos]
print(nea_onomata_comprehension)
````

Η σύνταξη αυτή ονομάζεται **list comprehension**: ένας τρόπος να δημιουργήσουμε μια λίστα σε μία μόνο γραμμή, με έναν βρόχο γραμμένο μέσα στις αγκύλες. Η μία αυτή γραμμή κάνει ακριβώς ό,τι και ο παραπάνω βρόχος, αλλά επιτρέπει στον προγραμματιστή να το εκφράσει με **πιο συμπαγή κώδικα** (tighter code), δηλαδή πιο συνοπτικό. Συχνά έτσι γράφεται ο προσεγμένος, τελικός κώδικας, και πρόκειται για ύφος που επιδιώκουν ιδιαίτερα οι επιστήμονες της πληροφορικής. Ο συμπαγής κώδικας θεωρείται συχνά πιο κομψός, μπορεί όμως να είναι δυσκολότερος στην κατανόηση για άλλους ερευνητές ή για αρχάριους προγραμματιστές. Όπως και στον γραπτό λόγο, ο κώδικας πρέπει να είναι ευανάγνωστος, και πρέπει να γνωρίζετε σε ποιους απευθύνεστε. Αν το κοινό σας δεν αποτελείται από επιστήμονες της πληροφορικής, ίσως είναι προτιμότερο να γράφετε πιο αναλυτικό κώδικα.

Ας αναλύσουμε τα συστατικά αυτής της γραμμής. Πρώτα έχουμε το αντικείμενο που θέλουμε να δημιουργήσουμε, με το όνομα μεταβλητής `nea_onomata_comprehension`, και το εξισώνουμε με μια αριστερή αγκύλη. Μέσα στις αγκύλες δηλώνουμε τι θέλουμε να κάνουμε με τη μεταβλητή που δημιουργείται προσωρινά σε κάθε επανάληψη του βρόχου, εδώ και πάλι την `onoma`: ορίζουμε ότι θέλουμε να ενώσουμε το δεύτερο κομμάτι του ονόματος με το πρώτο. Παρατηρήστε ότι εδώ η `split()` εκτελείται δύο φορές, γιατί δεν υπάρχει θέση για τη μεταβλητή `kommatia`: η μία γραμμή είναι πιο σύντομη, αλλά και πιο δύσκολη στην ανάγνωση. Στη συνέχεια ορίζουμε τον βρόχο, `for onoma in katalogos`. Τέλος, η δεξιά αγκύλη δηλώνει το τέλος της έκφρασης.

Προς το παρόν, προτιμήστε να είστε αναλυτικοί και να γράφετε τον κώδικά σας όπως στο πρώτο παράδειγμα. Έτσι θα κατανοήσετε καλύτερα τη λογική και θα αποσφαλματώνετε ευκολότερα τα προβλήματα που προκύπτουν, αφού κάθε σφάλμα θα εμφανίζεται στη συγκεκριμένη γραμμή που το προκάλεσε. Όταν εξοικειωθείτε περισσότερο με την Python, δοκιμάστε να χρησιμοποιήσετε list comprehension.

Παρουσιάζω εδώ το list comprehension όχι για να το χρησιμοποιήσετε, αλλά για να το δείτε πρώτα σε ένα ελεγχόμενο περιβάλλον. Είναι πιθανό να το συναντήσετε σε αποθετήρια του GitHub και στο StackOverflow χωρίς ιδιαίτερη εξήγηση. Τώρα που έχετε δει πώς μοιάζει ένα list comprehension και κατανοείτε τα συστατικά του, θα μπορείτε να αναλύετε τέτοιον κώδικα λίγο ευκολότερα.

## Ευρετηρίαση λίστας με και χωρίς enumerate

Συχνά, όταν γράφουμε έναν βρόχο, χρειάζεται να γνωρίζουμε σε ποια θέση της λίστας βρισκόμαστε. Όπως θα διαπιστώσετε, στον προγραμματισμό υπάρχουν πολλοί τρόποι να γίνει αυτό.

Ας δούμε πρώτα μια απλή προσέγγιση, ώστε να κατανοήσετε τη λογική της. Στη συνέχεια θα δούμε ένα πιο καθαρό, αλλά πιο σύνθετο παράδειγμα.

````python
i = 0
for onoma in katalogos:
    print(i)
    print(onoma)
    i=i+1
````

Στο παραπάνω παράδειγμα όλα είναι σχεδόν ίδια με τον αρχικό βρόχο for, με μία προφανή προσθήκη: τη μεταβλητή `i`. Το `i` είναι ένας μετρητής που ορίζουμε έξω από τον βρόχο, με αρχική τιμή 0. Στο τέλος κάθε επανάληψης εκτελούμε την εντολή `i=i+1`, η οποία λέει στην Python: «πάρε το αντικείμενο `i` και πρόσθεσε 1 στην τιμή του». Έτσι, σε κάθε πέρασμα από τον `katalogos` γνωρίζουμε ακριβώς σε ποια θέση βρισκόμαστε.

Τον ίδιο ακριβώς κώδικα μπορούμε όμως να τον γράψουμε με την ενσωματωμένη συνάρτηση `enumerate`. Η `enumerate` δημιουργεί αυτόματα για εμάς μια μεταβλητή-μετρητή κατά την εκτέλεση του βρόχου και την αυξάνει κάθε φορά που περνά σε επόμενο στοιχείο της λίστας. Με την `enumerate` γράφουμε πιο καθαρό και συμπαγή κώδικα.

````python
for i, onoma in enumerate(katalogos):
    print(i)
    print(onoma)
````

Παρατηρήστε ότι παίρνουμε ακριβώς το ίδιο αποτέλεσμα με παραπάνω. Ας αναλύσουμε τι συμβαίνει σε αυτό το παράδειγμα.

Το μόνο που έχει αλλάξει είναι η ακόλουθη γραμμή:

```
for i, onoma in enumerate(katalogos):
```

Παρατηρήστε την προσθήκη του `i, onoma`. Αυτό σημαίνει ότι σε κάθε επανάληψη του βρόχου δημιουργούνται δύο μεταβλητές. Όταν χρησιμοποιείτε την `enumerate`, η πρώτη μεταβλητή πρέπει πάντα να είναι το `i` ή κάποιο άλλο όνομα για τον ακέραιο-μετρητή. Ακολουθεί ένα κόμμα, που χωρίζει τις δύο μεταβλητές του βρόχου, και στη συνέχεια η μεταβλητή για το τρέχον στοιχείο της λίστας, με το ίδιο όνομα που χρησιμοποιήσαμε πριν, `onoma`. Τέλος, το `enumerate(katalogos):` λέει στην Python να εφαρμόσει την `enumerate` στον `katalogos`. Η συνάρτηση `enumerate` είναι αυτή που δημιουργεί για εμάς το `i` στη μνήμη.

Στο παρακάτω πλαίσιο, δημιουργήστε τη δική σας λίστα και διατρέξτε τη με έναν βρόχο. Στη συνέχεια, χρησιμοποιήστε την `enumerate` για να εκτυπώσετε τη θέση κάθε στοιχείου της λίστας.

## Τελεστές

Στο κεφάλαιο 1.2 γνωρίσαμε εν συντομία τους τελεστές (operators). Τους ανέφερα εκεί για να τους γνωρίζετε και επειδή συνήθως συνδέονται με αριθμούς· όπως θα δούμε, όμως, μπορούν να χρησιμοποιηθούν με όλους τους τύπους δεδομένων. Στους βρόχους θα χρειαστούμε πολύ συχνά τους τελεστές σύγκρισης (ίσο με, μικρότερο από κ.λπ.). Ακολουθεί ο κατάλογός τους:

1. Ίσο με (==)
2. Μεγαλύτερο από (>)
3. Μικρότερο από (<)
4. Μικρότερο ή ίσο με (<=)
5. Μεγαλύτερο ή ίσο με (>=)
6. Διάφορο του (!=)

Οι τελεστές σύγκρισης επιστρέφουν μια λογική τιμή (`True` ή `False`) ως απάντηση στο ερώτημα που θέτουμε με αυτούς. Ας υποθέσουμε, για παράδειγμα, ότι θέλουμε να μάθουμε αν το 1 είναι μικρότερο από το 2. Η απάντηση είναι προφανώς «αληθές»· στην Python το ερώτημα διατυπώνεται ως εξής:

````python
print (1 < 2)
````

Μπορούμε επίσης να διατυπώσουμε την ψευδή πρόταση ότι το 1 είναι μεγαλύτερο από το 2.

````python
print (1 > 2)
````

Ή ακόμη να ρωτήσουμε αν το 1 είναι ίσο με το 2. Παρατηρήστε το διπλό `==`. Χρειαζόμαστε δύο σύμβολα `=`, επειδή ένα μόνο `=` στην Python δημιουργεί μια νέα μεταβλητή.

````python
print (1 == 2)
````

Στα μαθηματικά, το «διάφορο του» δηλώνεται με ένα διαγραμμένο σύμβολο ισότητας (≠). Στον κώδικα αυτό δεν είναι πρακτικό, γιατί ο χαρακτήρας δεν υπάρχει στο πληκτρολόγιο· γι' αυτό χρησιμοποιούμε το `!=`. Είτε το πιστεύετε είτε όχι, με τον καιρό θα το γράφετε με μεγάλη ευκολία. Για να ελέγξουμε αν το 1 είναι διάφορο του 2, γράφουμε:

````python
print (1 != 2)
````

Εκ πρώτης όψεως, ίσως νομίζετε ότι δεν θα χρειαστεί ποτέ να χρησιμοποιήσετε τελεστές σύγκρισης· άλλωστε, ξέρετε ότι το 1 δεν είναι ίσο με το 2. Στην πραγματικότητα, θα τους χρησιμοποιείτε συνεχώς, επειδή σας επιτρέπουν να αξιοποιήσετε τη λογική Boolean, δηλαδή τη λογική του αληθούς και του ψευδούς. Αυτό είναι ιδιαίτερα χρήσιμο στους βρόχους, όπως θα δείξει ένας βρόχος while.

## Βρόχοι while

Οι τελεστές αυτοί μας επιτρέπουν να διατυπώσουμε σύνθετες συνθήκες μέσα στον βρόχο. Μπορούμε έτσι να πούμε ότι, όσο ισχύει κάτι, η Python πρέπει να εκτελεί κάποια ενέργεια. Θα συναντήσουμε την ίδια έννοια στην επόμενη ενότητα, όταν εξετάσουμε τις δηλώσεις συνθήκης (conditionals). Νομίζω ότι ο καλύτερος τρόπος να την κατανοήσετε είναι να τη δοκιμάσετε αμέσως στην πράξη.

Ο βρόχος for διατρέχει ένα σύνολο δεδομένων. Ο βρόχος while λειτουργεί λίγο διαφορετικά: εκτελείται επανειλημμένα όσο μια συνθήκη είναι αληθής. Όπως ο βρόχος for, ξεκινά με τη λέξη που του δίνει το όνομα, `while`. Στη συνέχεια δηλώνουμε τη συνθήκη που καθορίζει πότε θα σταματήσει ο βρόχος. Ας υποθέσουμε ότι θέλουμε να μετρήσουμε από το 0 έως το 10. Θα δημιουργούσαμε ένα αντικείμενο με το όνομα `x` και τιμή ίση με την αρχική τιμή. Στη συνέχεια θα δηλώναμε: όσο το `x != 10` (το x είναι διάφορο του 10), εκτύπωσε το `x`. Για να αυξάνεται το `x`, πρέπει σε κάθε επανάληψη να το αντικαθιστούμε με `x+1`, ώστε να μεγαλώνει κατά 1.

````python
x=0
while x != 10:
    print (x)
    x=x+1
````

Το έχω ήδη πει, αλλά αξίζει να το επαναλάβω: στον προγραμματισμό σπάνια υπάρχει μόνο ένας τρόπος για να εκτελεστεί μια εργασία. Τι διαφέρει στο παρακάτω μπλοκ κώδικα; Ρίξτε μια ματιά και δείτε αν μπορείτε να καταλάβετε τι ακριβώς συμβαίνει και γιατί επιτυγχάνει το ίδιο αποτέλεσμα.
