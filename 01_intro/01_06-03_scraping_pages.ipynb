# Scraping ιστοσελίδων με τις requests και BeautifulSoup

## Εισαγωγή

Στην προηγούμενη ενότητα γνωρίσαμε τα βασικά της HTML και τη βιβλιοθήκη BeautifulSoup. Δεν εργαστήκαμε όμως με δεδομένα από τον ιστό· η HTML μας ήταν αποθηκευμένη τοπικά. Σε αυτή την ενότητα θα μάθουμε πώς στέλνουμε αιτήματα σε έναν απομακρυσμένο διακομιστή με τη βιβλιοθήκη requests και πώς αναλύουμε στη συνέχεια τα δεδομένα με την BeautifulSoup.

## Requests

Η βιβλιοθήκη **requests** μάς επιτρέπει να στέλνουμε αιτήματα σε έναν διακομιστή μέσω της Python. Σκεφτείτε τη requests ως έναν αόρατο φυλλομετρητή που λειτουργεί στο παρασκήνιο του υπολογιστή σας. Κάνει ακριβώς ό,τι κάνει και ο φυλλομετρητής σας: στέλνει ένα αίτημα μέσω του διαδικτύου για να συνδεθεί σε μια συγκεκριμένη διεύθυνση διακομιστή. Κάθε διακομιστής έχει μια μοναδική διεύθυνση IP, αλλά συνήθως αντιστοιχεί σε αυτήν και μια μοναδική διεύθυνση με όνομα, μέσω της οποίας συνδεόμαστε στον διακομιστή χωρίς να χρειάζεται να πληκτρολογήσουμε τη διεύθυνση IP. Σε αντίθεση όμως με τον φυλλομετρητή σας, η requests δεν εμφανίζει τα αποτελέσματα· λαμβάνει την απάντηση και απλώς αποθηκεύει τα δεδομένα HTML στη μνήμη.

Για να δούμε πώς λειτουργεί, ας εισαγάγουμε πρώτα τη βιβλιοθήκη requests.

```python
import requests
```

Αφού εισαγάγαμε τη requests, ας δημιουργήσουμε ένα αντικείμενο συμβολοσειράς με τη διεύθυνση του ιστότοπου στον οποίο θέλουμε να κάνουμε scraping. Στον κώδικά μου ονομάζω πάντα αυτή τη συμβολοσειρά `url`.

```python
url = "https://en.wikipedia.org/wiki/List_of_French_monarchs"
```

Εξαιρετικά! Μπορούμε τώρα να χρησιμοποιήσουμε τη βιβλιοθήκη requests για να στείλουμε αίτημα στη συγκεκριμένη σελίδα. Αυτό γίνεται με τη συνάρτηση `get` της βιβλιοθήκης, η οποία έχει ένα υποχρεωτικό όρισμα: τη διεύθυνση της σελίδας που ζητάτε, εδώ τη συμβολοσειρά `url`.

```python
s = requests.get(url)
```

Αφού δημιουργήσαμε ένα αντικείμενο απάντησης (response object), ας δούμε πώς μοιάζει.

```python
print (s)
```

Εκ πρώτης όψεως, ίσως φαίνεται ότι αποτύχαμε. Τι είναι αυτό το παράξενο "Response [200]" και τι σημαίνει; Η συγκεκριμένη απάντηση σημαίνει ότι η σύνδεσή μας με τον διακομιστή ήταν επιτυχής. Υπάρχουν πολλοί κωδικοί απάντησης, αλλά το 200 είναι αυτός που θέλουμε να βλέπουμε. Αν λάβετε ποτέ απάντηση διαφορετική από 200, αναζητήστε τον συγκεκριμένο κωδικό στο Google για να μάθετε τι συμβαίνει. Μερικές φορές η απάντηση δείχνει ότι το αίτημά σας μπλοκαρίστηκε, ίσως επειδή ο ιστότοπος λαμβάνει μέτρα κατά του web scraping. Σε άλλες περιπτώσεις, η σελίδα μπορεί να είναι προστατευμένη, δηλαδή να απαιτεί σύνδεση (login). Τα πιθανά σφάλματα είναι πάρα πολλά για να περιγραφούν όλα σε αυτό το εισαγωγικό κεφάλαιο. Θα σας δώσω πάντως τη λύση σε ένα πολύ συνηθισμένο πρόβλημα, που μπορεί να σας βοηθήσει να ξεπεράσετε μια συχνή απάντηση 403. Για τη λύση αυτή, δείτε την τελευταία ενότητα του κεφαλαίου.

## BeautifulSoup

Αφού μάθαμε να στέλνουμε αίτημα σε έναν διακομιστή και αποθηκεύσαμε την απάντηση (την HTML) στη μνήμη, χρειαζόμαστε έναν τρόπο να αναλύσουμε αυτά τα δεδομένα. Το περιεχόμενο HTML βρίσκεται μέσα στο αντικείμενο απάντησης `s`, και η πρόσβαση σε αυτό γίνεται μέσω του χαρακτηριστικού `content`. Ας το προσπελάσουμε και ας εξετάσουμε τους πρώτους 100 χαρακτήρες.

```python
print (s.content[:100])
```

Παρατηρήστε ότι τα δεδομένα αυτά είναι HTML. Σε αυτό το στάδιο, ωστόσο, δεν έχουμε εύκολο τρόπο να επεξεργαστούμε αυτή τη συμβολοσειρά ως δομημένα δεδομένα. Εδώ μπαίνει στο παιχνίδι η BeautifulSoup, η οποία μετατρέπει το `s.content` σε δομημένα δεδομένα που μπορούμε στη συνέχεια να αναλύσουμε. Πρώτα πρέπει να εισαγάγουμε την BeautifulSoup. Σε αντίθεση με τις περισσότερες βιβλιοθήκες, η BeautifulSoup εισάγεται από το πακέτο bs4 (BeautifulSoup 4), οπότε εισάγουμε την κλάση `BeautifulSoup` από το `bs4`. Αυτό κάνει η παρακάτω εντολή.

```python
from bs4 import BeautifulSoup
```

Στη συνέχεια πρέπει να δημιουργήσουμε ένα νέο αντικείμενο soup.

```python
soup = BeautifulSoup(s.content)
```

Αν δεν εμφανιστεί σφάλμα, το αντικείμενο soup δημιουργήθηκε με επιτυχία. Ας το εκτυπώσουμε για να δούμε πώς μοιάζει:

```python
print (str(soup)[:200])
```

Το αντικείμενο soup μοιάζει ακριβώς ίδιο με το `s.content`, αλλά είναι εντελώς διαφορετικό. Διατηρεί τη δομή της HTML, επειδή η BeautifulSoup την έχει αναλύσει για εμάς, και έτσι μπορούμε να χρησιμοποιήσουμε ειδικές μεθόδους. Σε αυτή την ενότητα θα χρησιμοποιήσουμε τις `find` και `find_all`:

- `find`: βρίσκει την πρώτη εμφάνιση μιας ετικέτας και ανακτά την ετικέτα μαζί με όλα τα ένθετα στοιχεία της.
- `find_all`: επιστρέφει μια λίστα με όλες τις ετικέτες του ζητούμενου τύπου, μαζί με τα ένθετα στοιχεία τους.

Ας δούμε ένα βασικό παράδειγμα της μεθόδου `find`.

```python
div = soup.find("div")
```

Εδώ ανακτήσαμε την πρώτη ετικέτα `div` της σελίδας. Οι ετικέτες `div`, ωστόσο, είναι πολύ συνηθισμένες, επειδή αποτελούν ένα από τα βασικά δομικά στοιχεία της HTML. Ας δούμε πόσες υπάρχουν σε ολόκληρη τη σελίδα. Αυτό γίνεται με τη μέθοδο `find_all` και, στη συνέχεια, με τη μέτρηση του μήκους της λίστας μέσω της συνάρτησης `len`.

```python
all_divs = soup.find_all("div")
print (len(all_divs))
```

Επομένως, αν θέλαμε να ανακτήσουμε μια συγκεκριμένη `div` μόνο με τις `find` και `find_all`, θα έπρεπε να μετρήσουμε όλες τις ετικέτες `div`, να βρούμε τη σωστή θέση (index) και να ανακτήσουμε την ετικέτα από εκεί. Αυτό δεν θα λειτουργούσε σε μεγάλη κλίμακα, επειδή η θέση θα διέφερε από σελίδα σε σελίδα, ακόμη κι αν η δομή της HTML ήταν παρόμοια σε όλες τις σελίδες ενός ιστότοπου. Χρειαζόμαστε μια καλύτερη λύση, και εδώ αξιοποιούμε το δεύτερο, προαιρετικό όρισμα: η `find` μπορεί να δεχτεί επίσης ένα λεξικό, με το οποίο εξειδικεύουμε την αναζήτηση στο αντικείμενο soup.

Ας υποθέσουμε ότι θέλω να ανακτήσω το κύριο σώμα του άρθρου της Wikipedia. Αν επιθεωρήσω τη σελίδα, θα δω ότι μια συγκεκριμένη ετικέτα `div` περιέχει όλα τα δεδομένα του σώματος του άρθρου. Η ετικέτα αυτή έχει ένα μοναδικό χαρακτηριστικό `id` με τιμή "mw-content-text". Μπορώ επομένως να δώσω ως δεύτερο όρισμα ένα λεξικό με κλειδί το `id` και τιμή την τιμή του `id`. Έτσι η BeautifulSoup θα βρει την πρώτη ετικέτα `div` της οποίας το `id` είναι "mw-content-text". Ας το δούμε στον κώδικα.

```python
body = soup.find("div", {"id": "mw-content-text"})
```

Αφού ανακτήσαμε το σώμα του άρθρου, μπορούμε να εκτυπώσουμε το κείμενό του με το `.text`.

```python
print (body.text[:500])
```

Εξαιρετικά! Τι γίνεται όμως αν θέλουμε να ανακτήσουμε το κείμενο διατηρώντας τη δομή των παραγράφων; Αυτό γίνεται με αναζήτηση στο επίπεδο του `body`. Το αντικείμενο `body` που δημιουργήσαμε είναι κι αυτό ένα αντικείμενο soup, που διατηρεί τη δομή της HTML, αλλά περιέχει μόνο τα δεδομένα που βρίσκονται μέσα στη συγκεκριμένη `div`. Μπορούμε λοιπόν να εφαρμόσουμε την `find_all` στο αντικείμενο `body` για να βρούμε όλες τις παραγράφους, ως εξής:

```python
paragraphs = body.find_all("p")
```

Μπορούμε τώρα να διατρέξουμε τις παραγράφους. Ας το κάνουμε για τις πέντε πρώτες και ας εκτυπώσουμε το κείμενό τους.

```python
for paragraph in paragraphs[:5]:
    print (paragraph.text)
```
