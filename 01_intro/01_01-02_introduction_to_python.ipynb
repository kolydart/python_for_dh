# Εισαγωγή στην Python

## Γιατί οι ερευνητές των ανθρωπιστικών επιστημών χρειάζεται να μάθουν προγραμματισμό;

Πριν από οτιδήποτε άλλο, αξίζει να τεθεί ένα απλό ερώτημα: *γιατί οι ερευνητές των ανθρωπιστικών επιστημών χρειάζεται να μάθουν προγραμματισμό;* Για να απαντηθεί, ας εξεταστεί ένα ιδανικό έργο ψηφιακών ανθρωπιστικών σπουδών.

Σε ένα ιδανικό έργο ψηφιακών ανθρωπιστικών σπουδών, ένας ερευνητής (ή μια ομάδα ερευνητών) αξιοποιεί τη **γνώση του γνωστικού του αντικειμένου** (domain knowledge), δηλαδή την ειδίκευσή του στο πεδίο. Ως ειδικός, ο υπεύθυνος του έργου, γνωστός ως επιστημονικά υπεύθυνος (Principal Investigator, PI), μεταφέρει την ιδέα του στην ομάδα, η οποία αποτελείται από άλλους ερευνητές του ίδιου πεδίου και, συνήθως, από έναν τεχνικό υπεύθυνο. Ο τεχνικός υπεύθυνος αναλαμβάνει (ανάλογα με τη χρηματοδότηση) είτε ο ίδιος όλες τις τεχνικές πτυχές του έργου (ανάπτυξη ιστοσελίδων, προγραμματισμό, διαχείριση δεδομένων κ.λπ.) είτε τον συντονισμό μιας ομάδας που τις αναλαμβάνει. Αυτό είναι το ιδανικό σενάριο: κάθε μέλος της ομάδας κάνει αυτό που ξέρει καλύτερα.

Ένα τέτοιο έργο απαιτεί δύο πόρους που οι περισσότεροι ερευνητές διαθέτουν σε ανεπαρκή ποσότητα: χρηματοδότηση και χρόνο. Ο αυτάρκης ερευνητής, που αξιοποιεί τη γνώση του πεδίου του και ταυτόχρονα διαθέτει την τεχνική επάρκεια να εργαστεί στο έργο, περιορίζει δραστικά και τα δύο αυτά προβλήματα με διάφορους τρόπους. Για να γίνει κατανοητό πώς, πρέπει πρώτα να εξεταστεί ο τρόπος με τον οποίο χρηματοδοτούνται τα έργα ψηφιακών ανθρωπιστικών σπουδών.

## Τι είναι η Python;

Η Python είναι μια γλώσσα προγραμματισμού. Οι γλώσσες προγραμματισμού είναι τρόποι με τους οποίους εμείς, οι άνθρωποι, γράφουμε εντολές που στη συνέχεια εκτελεί ένας υπολογιστής. Οι ερευνητές μπορούν να επιλέξουν ανάμεσα σε πολλές γλώσσες προγραμματισμού, όπως:

1. C
2. Python
3. JavaScript
4. R
5. HTML (η κατάταξή της στις γλώσσες προγραμματισμού είναι αμφιλεγόμενη)

Δεν είναι όλες οι γλώσσες προγραμματισμού ισοδύναμες. Κάποιες ενδείκνυνται για την ανάπτυξη λογισμικού, όπως η C· άλλες είναι καταλληλότερες για την ανάπτυξη ιστοσελίδων, όπως η HTML και η JavaScript· και άλλες διακρίνονται στη στατιστική ανάλυση, όπως η R (και η Python). Πού υπερέχει η Python; Σε πολλούς τομείς. Αυτό που τη διαφοροποιεί από άλλες γλώσσες για όσους ξεκινούν τον προγραμματισμό είναι ότι διαβάζεται και γράφεται εύκολα.

Γράφεται σχετικά εύκολα σε σύγκριση με άλλες γλώσσες προγραμματισμού, επειδή η σύνταξή της (syntax), δηλαδή ο τρόπος με τον οποίο γράφονται οι εντολές στον κώδικα, είναι απλή και άμεση. Διαβάζεται εύκολα, επειδή η Python επιβάλλει τη χρήση εσοχών (indentation). Έτσι, τα μπλοκ κώδικα, που σε άλλες γλώσσες προγραμματισμού μπορεί να διαβάζονται δύσκολα, στην Python εντοπίζονται με μια ματιά.

Από τη δημιουργία της, στις αρχές της δεκαετίας του 1990, η δημοτικότητα της Python έχει αυξηθεί σημαντικά, γεγονός που οδήγησε σε μια μεγάλη κοινότητα προγραμματιστών και σε πλήθος διαθέσιμων βιβλιοθηκών. Οι βιβλιοθήκες θα εξεταστούν αργότερα στο εγχειρίδιο. Προς το παρόν, αρκεί να τις φανταστείτε ως μεγάλα σύνολα κώδικα που έχουν ήδη γραφτεί, ώστε με μία μόνο γραμμή κώδικα να εκτελείτε μια σύνθετη εργασία που διαφορετικά θα απαιτούσε εκατοντάδες ή χιλιάδες γραμμές.

Σήμερα η Python είναι μία από τις πιο διαδεδομένες γλώσσες προγραμματισμού και θεωρείται η κατεξοχήν γλώσσα για την ανάλυση κειμένου, τη μηχανική μάθηση, την ανάλυση δεδομένων (μαζί με την R), την εξαγωγή δεδομένων από τον ιστό (web scraping) και πολλά άλλα.

Κατά τη συγγραφή του βιβλίου, η τρέχουσα έκδοση της Python ήταν η `3.11.0`. Ας εξεταστεί καθένας από αυτούς τους αριθμούς. Το `3` αναφέρεται στην κύρια έκδοση της γλώσσας. Η Python 2 υπάρχει ακόμη (μάλιστα είναι προεγκατεστημένη σε όλους τους υπολογιστές Mac), αλλά δεν υποστηρίζεται πλέον και αντικαθίσταται σταδιακά από την Python 3. Δεν αξίζει να επενδύσετε χρόνο στην εκμάθηση της Python 2, καθώς χρησιμοποιείται μόνο για τη συντήρηση παλαιού (legacy) λογισμικού και, επειδή δεν συντηρείται πλέον, έχει πολλά ζητήματα ασφαλείας. Η διάκριση αυτή είναι σημαντική, γιατί ορισμένα πράγματα γράφονται διαφορετικά στις δύο εκδόσεις. Αν αναζητάτε βοήθεια σε ένα φόρουμ προγραμματισμού, όπως το StackOverflow, πρέπει να έχετε υπόψη σας αυτές τις διαφορές.

Το `11.0` στο `Python 3.11.0` δείχνει ποια ακριβώς έκδοση της Python 3 χρησιμοποιούμε. Ο κώδικας ορισμένων βιβλιοθηκών δεν είναι συμβατός με παλαιότερες εκδόσεις (backward compatible), δηλαδή κάποιες βιβλιοθήκες απαιτούν συγκεκριμένη έκδοση της Python· η κατανόηση αυτής της έννοιας από τώρα θα σας γλιτώσει από σύγχυση αργότερα. Κάθε νέα έκδοση φέρνει νέες δυνατότητες, οπότε είναι χρήσιμο να ενημερώνεστε για τις τρέχουσες εκδόσεις, χωρίς αυτό να είναι πάντα απαραίτητο.

Είναι από τις ευκολότερες στην εκμάθηση, γράφεται απλά και μπορεί να λύσει τα περισσότερα προγραμματιστικά προβλήματα που ενδέχεται να αντιμετωπίσει ένας ερευνητής.
