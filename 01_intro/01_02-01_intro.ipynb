2) Τύποι δεδομένων της Python<br>

5) Λογικές τιμές<br>
6) Χειρισμός συμβολοσειρών<br>

Το κεφάλαιο αυτό εξετάζει τα δεδομένα και τις δομές δεδομένων. Όπως σε κάθε γλώσσα προγραμματισμού, η βασική κατανόηση αυτών των δύο εννοιών είναι σημαντική, γιατί αποτελούν τα δομικά στοιχεία των περισσότερων πραγμάτων που θα κάνετε στην Python. Κάθε φορά που γράφετε ένα πρόγραμμα ή εκτελείτε μια εργασία με κώδικα, ουσιαστικά δίνετε εντολές για τη φόρτωση, την αποθήκευση και τον χειρισμό δεδομένων ή για την αλληλεπίδραση με αυτά. Η κατανόηση των διαφορετικών τύπων δεδομένων και του τρόπου με τον οποίο δομούνται και αποθηκεύονται είναι, επομένως, απαραίτητη.
