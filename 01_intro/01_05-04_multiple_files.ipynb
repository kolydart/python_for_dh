# Εργασία με πολλά αρχεία

## Εισαγωγή

Συχνά χρειάζεται να ανοίξετε πολλά αρχεία σε ένα σενάριο Python. Υπάρχουν πολλοί τρόποι να γίνει αυτό, αλλά, σε αντίθεση με τα περισσότερα εγχειρίδια Python, συνιστώ στους αρχάριους τη βιβλιοθήκη glob. Η <b>glob</b> περιλαμβάνεται στη βασική διανομή της Python, οπότε δεν χρειάζεται να την εγκαταστήσετε. Σκεφτείτε την ως μια βιβλιοθήκη που σας επιτρέπει να εξετάζετε τα περιεχόμενα ενός καταλόγου και να βρίσκετε όλα τα αρχεία που πληρούν ορισμένα κριτήρια.

## Εργασία με τη glob

Η glob μπορεί να σας μπερδέψει λίγο στην αρχή, οπότε ας την εξετάσουμε βήμα βήμα. Πρώτα πρέπει να την εισαγάγουμε.

```python
import glob, os
os.chdir("..")
```

Στη συνέχεια χρησιμοποιούμε τη συνάρτηση `glob.glob()`, η οποία δέχεται ένα όρισμα: μια συμβολοσειρά που περιγράφει τα αρχεία που θέλετε να βρείτε. Ας πάρουμε ως παράδειγμα τον υποφάκελο data. Στο παρακάτω παράδειγμα, η συμβολοσειρά αποτελείται από το όνομα του υποφακέλου με τα δεδομένα, μια κάθετο και έναν αστερίσκο (\*). Ο \* ονομάζεται χαρακτήρας υποκατάστασης (wildcard) και εδώ αντιστοιχεί σε όλα τα αρχεία του καταλόγου.

```python
files = glob.glob("data/*")
```

Ας εκτυπώσουμε τώρα τα αρχεία, για να δούμε τι περιέχει ο φάκελος.

```python
print (files)
```

Παρατηρήστε ότι συλλέξαμε όλα τα αρχεία! Τις περισσότερες φορές, ωστόσο, θα θέλετε μόνο αρχεία συγκεκριμένου τύπου, π.χ. .txt ή .json. Για τον σκοπό αυτό προσθέτουμε την κατάληξη .txt μετά τον \*, ώστε να συλλεχθούν όλα τα αρχεία .txt.

```python
files2 = glob.glob("data/*.txt")
print (files2)
```

## Αρχεία σε ένθετους καταλόγους

Αν κοιτάξετε στον υποκατάλογο data, θα δείτε δύο ένθετους καταλόγους, τους "other" και "other2". Μπορούμε να συλλέξουμε όλα τα αρχεία κάθε καταλόγου με τον ίδιο χαρακτήρα υποκατάστασης \*.

```python
files3 = glob.glob("data/*/*.txt")
print (files3)
```

## Διάσχιση καταλόγου

Η glob είναι εύκολη στη χρήση, έχει όμως ορισμένους περιορισμούς, για παράδειγμα όταν χρειάζεται να διασχίσετε έναν κατάλογο. Ας υποθέσουμε ότι θέλουμε να συλλέξουμε όλα τα αρχεία txt στους καταλόγους data, data/other και data/other2. Για να το κάνουμε, πρέπει να διασχίσουμε όλους τους υποκαταλόγους του data και να συλλέξουμε όλα τα σχετικά αρχεία. Αυτό δεν γίνεται με τη χρήση της glob που είδαμε μέχρι τώρα. Σε αυτές τις σπάνιες περιπτώσεις, πρέπει να είστε εξοικειωμένοι με τη βιβλιοθήκη os.

Η βιβλιοθήκη <b>os</b> μας επιτρέπει να κάνουμε στην Python πολλά πιο προχωρημένα πράγματα, τα οποία δεν θα καλύψω σε αυτό το εισαγωγικό εγχειρίδιο. Μία από τις κύριες χρήσεις της είναι η πλοήγηση σε καταλόγους και η δημιουργία καταλόγων. Οι χρήστες Linux θα βρουν μεγάλο μέρος της σύνταξης οικείο, ενώ στους χρήστες Windows ίσως φανεί κάπως ξένη. Προς το παρόν, ας εισαγάγουμε απλώς την os.

```python
import os
```

Αφού εισαγάγουμε την os, μπορούμε να χρησιμοποιήσουμε τη συνάρτηση `os.walk()`, η οποία δέχεται ως όρισμα μια συμβολοσειρά: τον κατάλογο από τον οποίο θέλετε να ξεκινήσει η διάσχιση. Ας υποθέσουμε ότι ο κατάλογός μας έχει την εξής δομή:

- data
    - other
    - other2

Θέλουμε να συλλέξουμε όλα τα αρχεία κειμένου στους καταλόγους data, other και other2.

```python
walking = os.walk("data/")
print (walking)
```

Η παραπάνω έξοδος μάς λέει ότι πρόκειται για μια γεννήτρια (generator). Οι γεννήτριες ξεφεύγουν από τους σκοπούς αυτού του εγχειριδίου· σκεφτείτε τες ως ένα ειδικό είδος αντικειμένου που υπάρχει στη μνήμη μόνο προσωρινά. Όποτε συναντάτε μια γεννήτρια, μπορείτε συνήθως να τη μετατρέψετε σε λίστα για να εργαστείτε μαζί της. Ας τη μετατρέψουμε με τη συνάρτηση `list()` της Python.

```python
walking = list(walking)
print (walking)
```

Το αποτέλεσμα δεν διαβάζεται εύκολα, οπότε ας διατρέξουμε ένα ένα τα στοιχεία του `walking`.

```python
for item in walking:
    print (item)
```

Όπως βλέπουμε, κάθε στοιχείο είναι μια πλειάδα με τρία μέρη:

- ριζικός κατάλογος (root)
- υποφάκελοι
- αρχεία

Για τους σκοπούς μας χρειαζόμαστε μόνο τον ριζικό κατάλογο και τα αρχεία. Ας τροποποιήσουμε τον βρόχο ώστε να ανακτά αυτά τα δύο δεδομένα και να τα εκτυπώνει.

```python
final_files = []
for item in walking:
    root = item[0]
    files = item[2]
    print ("This is the Root")
    print (root)
    print ("These are the Files")
    print (files)
    print ()
    print ()
```

Όπως βλέπουμε, τα αρχεία αποτελούν μια λίστα. Μπορούμε επομένως να διατρέξουμε τη λίστα των αρχείων και να ενώσουμε τον ριζικό κατάλογο με κάθε όνομα αρχείου, ώστε να σχηματίσουμε τις πλήρεις διαδρομές. Στην os αυτό γίνεται με τη συνάρτηση `os.path.join()`, η οποία δέχεται δύο ορίσματα: τον ριζικό κατάλογο και το αρχείο. Ας τροποποιήσουμε ξανά τον βρόχο, κρατώντας μόνο τα αρχεία .txt, και ας εκτυπώσουμε τα αποτελέσματα.

```python
for item in walking:
    root = item[0]
    files = item[2]
    for file in files:
        if file.endswith(".txt"):
             print(os.path.join(root, file))
```

Εξαιρετικά! Μπορούμε τώρα να χρησιμοποιήσουμε τον ίδιο ακριβώς βρόχο για να προσθέσουμε τις διαδρομές των αρχείων σε μια αρχικά κενή λίστα με το όνομα `final_files`.

```python
final_files = []
for item in walking:
    root = item[0]
    files = item[2]
    for file in files:
        if file.endswith(".txt"):
             final_files.append(os.path.join(root, file))
print (final_files)
```

Παρατηρήστε ότι έχουμε πλέον όλα τα αρχεία .txt του κύριου καταλόγου και όλων των υποκαταλόγων του. Αυτό θα σας φανεί πολύ χρήσιμο όταν τα αρχεία σας είναι κατανεμημένα σε πολλά επίπεδα υποκαταλόγων.
