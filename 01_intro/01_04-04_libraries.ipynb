# Βιβλιοθήκες στην Python

## Εισαγωγή

Οι βιβλιοθήκες (libraries) υπάρχουν σε όλες τις γλώσσες προγραμματισμού. Σας επιτρέπουν να εισαγάγετε μεγάλες ποσότητες κώδικα, με συναρτήσεις και κλάσεις που μπορείτε να αξιοποιήσετε στον δικό σας κώδικα. Η λογική των βιβλιοθηκών συνοψίζεται στην παλιά φράση «μην ανακαλύπτεις ξανά τον τροχό». Χρησιμοποιήστε ό,τι έχουν ήδη φτιάξει άλλοι! Ο ανοιχτός κώδικας είναι ανοιχτός για κάποιο λόγο! Όσοι δημιουργούν αυτές τις βιβλιοθήκες το κάνουν ώστε να μη χρειάζεται να λύσετε εσείς συγκεκριμένα προβλήματα. Τα έχουν ήδη λύσει για εσάς.

Σε αυτό το κεφάλαιο θα μάθουμε πώς εγκαθιστούμε και εισάγουμε βιβλιοθήκες. Το κάνουμε σε αυτό το σημείο επειδή στη συνέχεια του εγχειριδίου θα χρειαστούμε εξωτερικές βιβλιοθήκες, συγκεκριμένα τις pandas, requests και BeautifulSoup.

## Πώς εγκαθίστανται οι βιβλιοθήκες της Python

Η Python διαθέτει προεγκατεστημένο το pip. Το <b>pip</b> είναι ένας διαχειριστής πακέτων (package manager): κατεβάζει, εγκαθιστά και διαχειρίζεται διαφορετικές εκδόσεις λογισμικού στον υπολογιστή σας. Αν χρησιμοποιείτε Windows, η έννοια αυτή ίσως σας είναι κάπως ξένη. Σκεφτείτε το pip ως ένα εργαλείο που διαχειρίζεται όλες τις βιβλιοθήκες σας για λογαριασμό σας. Αν εγκαταστήσατε την Python μέσω του Anaconda, όπως σύστησα, το pip θα προστεθεί αυτόματα στη μεταβλητή Path του συστήματός σας. Η <b>Path</b> στα Windows είναι ο μηχανισμός μέσω του οποίου ο υπολογιστής αναγνωρίζει εντολές που αντιστοιχούν σε εκτελέσιμα αρχεία (exe) του συστήματος. Για να το ελέγξετε, ανοίξτε το τερματικό και πληκτρολογήστε `pip --version` (στα Windows, το τερματικό είναι η γραμμή εντολών, command prompt).

Επειδή σε αυτό το εγχειρίδιο χρησιμοποιώ το Jupyter Notebook, μπορώ να εκτελώ εντολές τερματικού γράφοντας `!` μπροστά από την εντολή. Όταν πληκτρολογήσετε `pip --version`, θα πρέπει να δείτε μια έξοδο παρόμοια με την ακόλουθη.

```python
!pip --version
```

## Πώς εισάγεται μια βιβλιοθήκη

Για να εισαγάγουμε μια βιβλιοθήκη, χρησιμοποιούμε την εντολή `import library_name`, ως εξής:

```python
import requests
```

Για ορισμένες βιβλιοθήκες υπάρχει ένας καθιερωμένος, Pythonic τρόπος εισαγωγής. Κλασική περίπτωση είναι η pandas, την οποία εισάγουμε `as pd`. Το `as pd` σημαίνει ότι μέσα στο σενάριο η βιβλιοθήκη δεν θα καλείται "pandas", αλλά "pd".

```python
import pandas as pd
```
