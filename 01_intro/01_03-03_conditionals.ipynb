# Δηλώσεις συνθήκης

Όπως οι βρόχοι, έτσι και οι δηλώσεις συνθήκης είναι απαραίτητο στοιχείο όλων των γλωσσών προγραμματισμού. Οι <b>δηλώσεις συνθήκης</b> (conditionals) είναι ένας τύπος λογικής στον προγραμματισμό που σας επιτρέπει να καθορίζετε αν θα εκτελεστεί κάτι, ανάλογα με το αν μια συνθήκη είναι αληθής ή ψευδής. Οι δηλώσεις συνθήκης είναι πάντα δυαδικές. Μου είναι πάντα χρήσιμο να τις διατυπώνω πρώτα σε φυσική γλώσσα.

Αν κάτι είναι αληθές, κάνε αυτό. Αν δεν είναι αληθές, μπορείτε να ορίσετε τι πρέπει να γίνει σε αυτή την περίπτωση. Σε ψευδοκώδικα, αυτό θα έμοιαζε κάπως έτσι:

```
αν κάτι είναι αληθές:
    κάνε αυτό
αλλιώς:
    κάνε κάτι άλλο
```
    
Παρατηρήστε και πάλι την εσοχή. Σε αυτό το κεφάλαιο θα γνωρίσετε τα τρία βασικά στοιχεία των δηλώσεων συνθήκης:


`````{admonition} Δηλώσεις συνθήκης
:class: hint
| κώδικας | περιγραφή                              |
|---------|----------------------------------------|
| `if`    | Ξεκινά μια συνθήκη («αν»).             |
| `else`  | Λειτουργεί ως «αλλιώς».                |
| `elif`  | Λειτουργεί ως «αλλιώς, αν».            |
`````

## Η δήλωση if

Ας ξεκινήσουμε με τη δήλωση `if`, δημιουργώντας πρώτα ένα αντικείμενο με το όνομα `x` και τιμή 0. Μπορούμε τώρα να πούμε: «αν το x είναι ίσο με 0, εκτύπωσέ το». Παρατηρήστε στον παρακάτω κώδικα την `:` και την εσοχή που ακολουθεί. Όπως είδαμε στους βρόχους, η `:` δηλώνει ότι η επόμενη γραμμή πρέπει να έχει εσοχή. Κάθε γραμμή κώδικα που περιέχει δήλωση συνθήκης πρέπει πάντα να τελειώνει με `:`, ώστε η Python να γνωρίζει πού τελειώνει η συνθήκη. Η επόμενη γραμμή πρέπει επίσης πάντα να έχει εσοχή.

``````python
x=0
if x==0:
    print (x)
``````

Η μεταβλητή `x` εκτυπώθηκε με επιτυχία, επειδή η συνθήκη που ορίσαμε είναι αληθής (`True`). Ας δοκιμάσουμε το αντίθετο, ορίζοντας τη συνθήκη `x == 1`.

``````python
if x==1:
    print (x)
``````

Παρατηρήστε ότι δεν συμβαίνει τίποτα, επειδή η συνθήκη μας είναι ψευδής (`False`). Στην παρακάτω εφαρμογή Trinket, δοκιμάστε να γράψετε μια δήλωση συνθήκης.

## Η δήλωση else

Αν θέλουμε να γίνει κάτι όταν η συνθήκη δεν ικανοποιείται, χρησιμοποιούμε τη δήλωση `else`. Σκεφτείτε την `else` ως «αλλιώς». Σε ψευδοκώδικα, η `else` θα έμοιαζε κάπως έτσι:

```
αν αυτή η συνθήκη είναι αληθής:
    κάνε αυτό
αλλιώς (else):
    κάνε κάτι άλλο
```

Ας το γράψουμε σε πραγματικό κώδικα.

``````python
if x==1:
    print (x)
else:
    print ("X is not 1")
``````

Παρατηρήστε ότι η `else` έχει την ίδια εσοχή με τη δήλωση `if` και ότι κι αυτή ακολουθείται από `:` και κώδικα με εσοχή. Όπως σημειώθηκε παραπάνω, όλες οι γραμμές με δήλωση συνθήκης πρέπει να τελειώνουν με άνω και κάτω τελεία, και η επόμενη γραμμή να έχει εσοχή.

Η δήλωση `else` είναι σημαντική, επειδή επιτρέπει στον προγραμματιστή να διατυπώσει πιο σύνθετη λογική.

## Η δήλωση elif και ο τελεστής in με συμβολοσειρές

Δίστασα να συμπεριλάβω την `elif` σε αυτό το κεφάλαιο, από φόβο μήπως εισαγάγω πολλά νέα στοιχεία ταυτόχρονα, αλλά θεωρώ σημαντικό να γνωρίζετε τουλάχιστον την ύπαρξή της, ακόμη κι αν δεν τη χρησιμοποιήσετε αμέσως. Η `elif` (συντομογραφία του else if) αντιστοιχεί περίπου στο «αλλιώς, αν».

```
αν η συνθήκη 1 είναι αληθής:
    κάνε αυτό
αλλιώς, αν (elif) η συνθήκη 2 είναι αληθής:
    κάνε κάτι άλλο
```

Όπως φαίνεται στο παράδειγμα, η `elif` λειτουργεί λίγο διαφορετικά από την `else`. Με την `else` εκτελούμε μια ενέργεια όταν η προηγούμενη συνθήκη είναι ψευδής· με την `elif` ορίζουμε ότι πρέπει, επιπλέον, να ισχύει μια άλλη συνθήκη.

Εκ πρώτης όψεως, η `elif` μοιάζει να λειτουργεί όπως δύο διαδοχικές δηλώσεις `if`, αλλά η συμπεριφορά της διαφέρει ελαφρώς. Για να καταλάβετε πώς, είναι καλύτερο να τη δείτε σε κώδικα.

Στον παρακάτω κώδικα, μέσα στη δήλωση συνθήκης θα δείτε τη λέξη `in`, η οποία είναι τελεστής. Στην Python, ο `in` λειτουργεί όπως η αγγλική λέξη in («μέσα σε»): ελέγχει αν κάτι περιέχεται σε ένα δεδομένο ή σε μια δομή δεδομένων. Αν, λοιπόν, θέλουμε να μάθουμε αν μια συμβολοσειρά περιέχεται σε μια άλλη, μπορούμε να χρησιμοποιήσουμε τον `in`.

Στο παρακάτω παράδειγμα θέλουμε να μάθουμε αν μια υποσυμβολοσειρά (μια μικρότερη συμβολοσειρά) εμφανίζεται μέσα στη μεταβλητή `text`.

``````python
text = "Στη συναυλία ακούστηκαν έργα του Σκαλκώτα και του Καλομοίρη."
if "Σκαλκώτα" in text:
    print("Βρέθηκε ο Σκαλκώτας")
elif "Καλομοίρη" in text:
    print("Βρέθηκε ο Καλομοίρης")
``````

Η έξοδος αυτή ίσως φαίνεται παράξενη. Και οι δύο συνθέτες αναφέρονται στη συμβολοσειρά `text`. Γιατί, λοιπόν, δεν εκτυπώνεται «Βρέθηκε ο Καλομοίρης»; Επειδή η `elif` εκτελείται μόνο αν δεν έχει ικανοποιηθεί καμία από τις προηγούμενες δηλώσεις `if` ή `elif`. Αν ήθελα να ελέγξω και τις δύο συνθήκες, θα χρησιμοποιούσα αντί γι' αυτό δύο δηλώσεις `if`, ως εξής:

``````python
if "Σκαλκώτα" in text:
    print("Βρέθηκε ο Σκαλκώτας")
if "Καλομοίρη" in text:
    print("Βρέθηκε ο Καλομοίρης")
``````

Η δήλωση `elif` επιτρέπει πιο σύνθετη λογική και διαδοχικούς ελέγχους συνθηκών. Στα πρώτα σας βήματα στον προγραμματισμό ίσως δεν τη χρησιμοποιείτε συχνά, είναι όμως σημαντικό να είστε εξοικειωμένοι μαζί της από νωρίς.

## Δηλώσεις συνθήκης και λίστες με in και not in

Οι δηλώσεις συνθήκης είναι συχνά απαραίτητες όταν ελέγχουμε αν κάτι περιέχεται σε μια λίστα. Θυμηθείτε ότι οι λίστες είναι δομές δεδομένων που περιέχουν μια σειρά δεδομένων. Τα στοιχεία μιας λίστας μπορεί να είναι συμβολοσειρές, ακέραιοι, δεκαδικοί ή ακόμη και άλλες δομές δεδομένων, όπως λίστες, πλειάδες και λεξικά. Συχνά θα χρειαστεί να ελέγξετε τι περιέχει μια λίστα. Για τον σκοπό αυτό, μπορούμε να χρησιμοποιήσουμε δηλώσεις συνθήκης με τον ίδιο τελεστή `in` που είδαμε παραπάνω.

Ας δημιουργήσουμε πρώτα μια λίστα με το όνομα `synthetes`.

``````python
synthetes = ["Σαμάρας", "Καλομοίρης", "Σκαλκώτας"]
``````

Τώρα που έχουμε μια λίστα ονομάτων, ας χρησιμοποιήσουμε τον τελεστή `in` για να ελέγξουμε αν το όνομα "Καλομοίρης" περιέχεται στη λίστα `synthetes`.

```{warning}
Θυμηθείτε: η Python κάνει διάκριση πεζών και κεφαλαίων.
```

Σε ψευδοκώδικα, θα μπορούσαμε να γράψουμε:

```
αν ο Καλομοίρης είναι στους synthetes:
    εκτύπωσε True
```

Επειδή η σύνταξη της Python μοιάζει πολύ με την αγγλική γλώσσα, αρκούν λίγες τροποποιήσεις για να γίνει ο ψευδοκώδικας πραγματικός κώδικας.

``````python
if "Καλομοίρης" in synthetes:
    print(True)
``````

Το ίδιο μπορεί να γίνει και αντίστροφα. Τι γίνεται αν θέλουμε να μάθουμε αν κάποιος άλλος συνθέτης δεν περιέχεται στη λίστα `synthetes`; Σε ψευδοκώδικα, αυτό θα έμοιαζε κάπως έτσι:

```
αν ο Ξενάκης δεν είναι στους synthetes:
    εκτύπωσε «ο Ξενάκης δεν είναι στη λίστα»
```

Η Python κάνει αυτή την κατασκευή απλή, επειδή ο τελεστής `in` έχει και αρνητική εκδοχή, τον `not in`. Ο τελεστής αυτός λειτουργεί με τον ίδιο ακριβώς τρόπο, αλλά δίνει το αντίθετο αποτέλεσμα από τον `in`. Ας τον δούμε στην πράξη.

``````python
if "Ξενάκης" not in synthetes:
    print("Ο Ξενάκης δεν είναι στη λίστα.")
``````

Παρατηρήστε ότι γράψαμε με επιτυχία μια δήλωση συνθήκης που ελέγχει αν μια συμβολοσειρά δεν περιέχεται στη λίστα `synthetes`. Οι `in` και `not in` ελέγχουν επίσης αν μια υποσυμβολοσειρά εμφανίζεται μέσα σε μια συμβολοσειρά. Για να ελέγξουμε αν η υποσυμβολοσειρά `Σκαλκώτας` εμφανίζεται στο κείμενο `Σκαλκώτας, Νίκος`, θα γράφαμε μια δήλωση συνθήκης όπως η εξής:

``````python
if "Σκαλκώτας" in "Σκαλκώτας, Νίκος":
    print("Βρέθηκε ο Σκαλκώτας.")
``````
