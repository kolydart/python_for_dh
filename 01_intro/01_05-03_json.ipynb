# Εργασία με δεδομένα JSON

## Εισαγωγή

Σε αυτό το κεφάλαιο θα γνωρίσετε τα δεδομένα JSON. Το JSON (JavaScript Object Notation) είναι μια μορφή δεδομένων που χρησιμοποιείται ευρέως στο διαδίκτυο. Είναι ο κύριος τρόπος με τον οποίο αποθηκεύονται δεδομένα για ιστότοπους και ανακτώνται από τη JavaScript. Υπάρχουν δύο λόγοι για να μάθετε τις δομές δεδομένων JSON νωρίς στην πορεία σας στον προγραμματισμό:

1) Το JSON αναγνωρίζεται από όλους τους φυλλομετρητές (browsers).
2) Το JSON σας επιτρέπει να δομείτε ιεραρχικά δεδομένα.

Τι είναι τα ιεραρχικά δεδομένα; Είναι δεδομένα που μπορεί να είναι ένθετα μέσα σε άλλα δεδομένα. Τέτοια δεδομένα συχνά δύσκολα αναπαρίστανται καθαρά σε μορφή CSV. Ας το δούμε με ένα παράδειγμα. Φανταστείτε ότι θέλετε να αποθηκεύσετε μια σειρά κειμένων στο Excel. Σε μια στήλη θα είχατε το κείμενο και στην επόμενη μια λίστα με τα πρόσωπα που αναφέρονται σε αυτό. Πώς όμως αναπαριστάνετε τη λίστα ονομάτων; Θα μπορούσατε να το κάνετε έτσι:

| Κείμενο                                                             | Ονόματα                        |
|------------------------------------------------------------------|------------------------------|
| Ο John και η Kathy γνωρίζουν ο ένας τον άλλον. Η Marge και ο Francois όχι. | John, Kathy, Marge, Francois |

Αυτό όμως γίνεται πολύ γρήγορα χαοτικό, επειδή μια λίστα δύσκολα αποθηκεύεται ως δομή δεδομένων μέσα σε ένα κελί. Υπάρχουν τρόποι να παρακαμφθεί το πρόβλημα, αλλά όταν επεξεργάζεστε υπολογιστικά τέτοια δεδομένα CSV, τα πράγματα περιπλέκονται όσο βαθαίνουν οι ιεραρχίες. Φανταστείτε ότι για κάθε πρόσωπο είχαμε επίσης ηλικία και ρόλο. Πώς θα αποθηκεύαμε αυτά τα δεδομένα; Ίσως κάπως έτσι:

| Κείμενο                                                             | Ονόματα                        | Ηλικία            |
|------------------------------------------------------------------|------------------------------|----------------|
| Ο John και η Kathy γνωρίζουν ο ένας τον άλλον. Η Marge και ο Francois όχι. | John, Kathy, Marge, Francois | 20, 25, 30, 35 |

Αν θέλω να ενημερώσω τα δεδομένα, πρέπει να βεβαιωθώ ότι η λίστα ηλικιών αντιστοιχεί στη λίστα ονομάτων, κάτι που μπορεί και πάλι να προκαλέσει προβλήματα στο μέλλον. Η λύση είναι να μη χρησιμοποιούμε CSV ή Excel για τέτοια δεδομένα, αλλά μια πιο ευέλικτη μορφή, ικανή να χειρίζεται λίστες και ένθετα ιεραρχικά δεδομένα. Στην Python, η απλούστερη λύση είναι το JSON. Για να το χρησιμοποιήσετε, δεν χρειάζεται να εγκαταστήσετε κάποια ειδική βιβλιοθήκη, γιατί η Python διαθέτει ενσωματωμένη τη βιβλιοθήκη `json`. Οι μόνες μέθοδοι που χρειάζεται να γνωρίζετε για να δουλέψετε με JSON είναι οι `json.dump()` και `json.load()`. Πρώτα όμως, ας εισαγάγουμε τη βιβλιοθήκη `json`.

```python
import json
```

Ας αναπαραστήσουμε τα παραπάνω δεδομένα ως δομή δεδομένων στην Python.

```python
data = {"text": "John and Kathy both know each other. Marge and Francois do not.",
        "names": ["John", "Kathy", "Marge", "Francois"]}
data
```

## json.dump()

Ας αποθηκεύσουμε τώρα αυτά τα δεδομένα έξω από την Python, σε ένα αρχείο JSON. Για τον σκοπό αυτό θα χρησιμοποιήσουμε την εντολή `with open` που μάθαμε στο προηγούμενο κεφάλαιο, δίνοντας όμως στο αρχείο την κατάληξη .json αντί για .txt. Στη συνέχεια θα εκτελέσουμε την `json.dump()`, η οποία δέχεται δύο βασικά ορίσματα: τα δεδομένα που θέλετε να γράψετε στο αρχείο και το αντικείμενο αρχείου στο οποίο θα γραφτούν, εδώ το `f`. Το πρόσθετο όρισμα λέξης-κλειδιού `indent` το χρησιμοποιώ πάντα, επειδή κάνει το αρχείο JSON πιο ευανάγνωστο: προσθέτει εσοχή 4 κενών σε κάθε βαθύτερο επίπεδο της ιεραρχίας.

```python
with open ("data/sample_json.json", "w") as f:
    json.dump(data, f, indent=4)
```

## json.load()

Αφού γράψαμε τα δεδομένα σε ένα αρχείο, ας τα φορτώσουμε ξανά. Θα ανοίξουμε το ίδιο αρχείο JSON, αυτή τη φορά για ανάγνωση, θα δημιουργήσουμε ένα νέο αντικείμενο, το `new_data`, και θα χρησιμοποιήσουμε την `json.load()`. Η μέθοδος δέχεται ένα όρισμα, το αντικείμενο αρχείου από το οποίο θέλετε να φορτώσετε τα δεδομένα. Εφόσον το αρχείο JSON δεν είναι κατεστραμμένο, τα δεδομένα θα φορτωθούν με επιτυχία.

```python
with open ("data/sample_json.json", "r") as f:
    new_data = json.load(f)
new_data
```

Αυτές είναι οι μόνες δύο εντολές που χρειάζεται να γνωρίζετε για να αρχίσετε να εργάζεστε με δεδομένα JSON στην Python. Σας ενθαρρύνω να αφιερώσετε μερικά λεπτά σε πειραματισμούς, αποθηκεύοντας δεδομένα με την `json.dump()` και φορτώνοντάς τα με την `json.load()`.
