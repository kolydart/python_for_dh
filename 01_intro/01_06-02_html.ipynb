# Εισαγωγή στην HTML

## Εισαγωγή

Σε αυτό το κεφάλαιο θα γνωρίσουμε το web scraping (συλλογή δεδομένων από τον ιστό), μία από τις πιο κρίσιμες δεξιότητες για έναν ερευνητή των ψηφιακών ανθρωπιστικών σπουδών. Το **web scraping** είναι η διαδικασία με την οποία αυτοματοποιούμε την αποστολή αιτήματος σε έναν διακομιστή (server), που φιλοξενεί έναν ιστότοπο, και την ανάλυση της απάντησης, δηλαδή ενός αρχείου HTML. Η **HTML** (HyperText Markup Language, Γλώσσα Σήμανσης Υπερκειμένου) είναι ο τρόπος με τον οποίο δομούνται οι ιστότοποι. Όταν κάνουμε scraping σε έναν ιστότοπο, διατυπώνουμε κανόνες για την εξαγωγή πληροφοριών με βάση τον τρόπο που είναι δομημένα τα δεδομένα μέσα στην HTML. Για να γίνουμε ικανοί στο web scraping, επομένως, πρέπει να μπορούμε να κατανοούμε και να αναλύουμε την HTML.

Σε αυτή την ενότητα θα αναλύσουμε την HTML και θα γνωρίσουμε τις πιο συνηθισμένες ετικέτες (tags), όπως οι div, p, strong και span. Θα γνωρίσουμε επίσης τα χαρακτηριστικά (attributes) αυτών των ετικετών, όπως τα href, class και id. Είναι πολύ σημαντικό να τα κατανοήσετε πριν προχωρήσετε στο επόμενο κεφάλαιο, όπου μαθαίνουμε πώς να κάνουμε web scraping με την Python.

## Εμβάθυνση στην HTML

Γιατί, λοιπόν, είναι χρήσιμη η HTML; Η HTML, όπως και άλλες γλώσσες σήμανσης, π.χ. η **XML** (eXtensible Markup Language, Επεκτάσιμη Γλώσσα Σήμανσης), επιτρέπει στους χρήστες να δομούν δεδομένα μέσα σε άλλα δεδομένα. Αυτό επιτυγχάνεται με τις λεγόμενες ετικέτες (tags). Είναι καλύτερο να το δούμε στην πράξη. Ας εξετάσουμε ένα απλό αρχείο HTML.

```
<div>
    <p>This is a paragraph</p>
</div>
```

Παραπάνω βλέπουμε τη δομή ενός πολύ απλού αρχείου HTML. Στην πρώτη γραμμή βλέπουμε το `<div>`. Παρατηρήστε τη χρήση των `<` και `>`. Το αριστερό `<` δηλώνει την έναρξη μιας ετικέτας. Η **ετικέτα** (tag) είναι ο τρόπος με τον οποίο δηλώνεται η δομή μέσα σε ένα αρχείο HTML: δείχνει ότι το περιεχόμενο που περικλείει ανήκει σε έναν συγκεκριμένο τύπο. Μετά το `<` βλέπουμε τη λέξη `div`, που δηλώνει τον τύπο της ετικέτας. Εδώ δημιουργούμε μια ετικέτα `div`, έναν από τους πιο συνηθισμένους τύπους ετικετών στην HTML. Μετά το όνομα της ετικέτας ακολουθεί το `>`, που σηματοδοτεί το τέλος της ετικέτας.

Στη γραμμή 2 βλέπουμε ένα ένθετο τμήμα HTML, με εσοχή. Στην HTML, σε αντίθεση με την Python, η εσοχή είναι προαιρετική. Ωστόσο, είναι καλή πρακτική να χρησιμοποιούμε αλλαγές γραμμής και εσοχές, ώστε το έγγραφο να διαβάζεται ευκολότερα από ανθρώπους. Η δεύτερη γραμμή αρχίζει με μια ετικέτα `p`, που στην HTML δηλώνει την αρχή μιας παραγράφου.

Μετά την ετικέτα `p` βλέπουμε το `This is a paragraph` («Αυτή είναι μια παράγραφος»). Στην HTML, το κείμενο που δεν βρίσκεται μέσα στα σύμβολα `< >` των ετικετών είναι το κείμενο που εμφανίζεται στην ιστοσελίδα. Εδώ, το αρχείο HTML θα εμφάνιζε το κείμενο `This is a paragraph`. Αμέσως μετά το κείμενο συναντάμε την πρώτη ετικέτα κλεισίματος (close tag), η οποία δηλώνει ότι το ένθετο τμήμα της δομής ολοκληρώθηκε. Στην περίπτωσή μας, η πρώτη ετικέτα κλεισίματος είναι η `</p>`. Την αναγνωρίζουμε από το `</` στη θέση του `<`.

Στην τελευταία γραμμή βλέπουμε την ετικέτα κλεισίματος της `div`.

## Κατανόηση των χαρακτηριστικών (attributes)

Ας δούμε ένα άλλο μπλοκ HTML, αυτή τη φορά με μια μικρή αλλαγή. Μπορείτε να την εντοπίσετε;

```
<div class="content">
    <p>This is a paragraph</p>
</div>
```

Αν απαντήσατε το τμήμα `class="content"` στην ετικέτα ανοίγματος της `div`, έχετε δίκιο. Το τμήμα αυτό βρίσκεται μέσα στην ετικέτα και ονομάζεται **χαρακτηριστικό** (attribute). Εδώ χρησιμοποιείται το χαρακτηριστικό `class` (ένας πολύ συνηθισμένος τύπος χαρακτηριστικού), με τιμή `content`.

Όταν κάνετε scraping σε ιστότοπους, μπορείτε να χρησιμοποιείτε τα χαρακτηριστικά για να ορίσετε ποια ετικέτα `div` θα ανακτήσετε. Τα πιο συνηθισμένα χαρακτηριστικά είναι τα `class` και `id`.

## Ανάλυση HTML με την BeautifulSoup

Αφού γνωρίσαμε λίγο την HTML, ας αρχίσουμε να την αναλύουμε με την Python. Όταν **αναλύουμε** (parse) HTML, αυτοματοποιούμε την αναγνώριση της δομής της και την ερμηνεύουμε συστηματικά. Αυτή είναι η βάση του web scraping. Για αρχή, ας δημιουργήσουμε ένα απλό αρχείο HTML στη μνήμη.

````python
html = """
<html>
    <body>
        <div class="content">
            <p>This is our first content</p>
        </div>
        <div class="other">
            <p>This is is another piece of content</p>
        </div>
    </body>
</html>
"""
````

Για την ανάλυση δεδομένων HTML υπάρχουν πολλές βιβλιοθήκες Python. Για πιο απαιτητικά προβλήματα, το [Selenium](https://selenium-python.readthedocs.io) αποτελεί το πρότυπο της βιομηχανίας. Το Selenium είναι ισχυρό, αλλά απαιτεί αρκετό χρόνο εκμάθησης και μπορεί να δυσκολέψει όσους μόλις ξεκινούν με την Python, ιδίως όσους προγραμματίζουν σε Windows. Επιπλέον, τα περισσότερα προβλήματα web scraping λύνονται και χωρίς το Selenium.

Για αυτούς τους λόγους, σε αυτό το εγχειρίδιο δεν θα χρησιμοποιήσουμε το Selenium, αλλά την [BeautifulSoup](https://pypi.org/project/beautifulsoup4/), μια ελαφριά, γρήγορη και αποτελεσματική βιβλιοθήκη Python για την ανάλυση HTML. Το δυσκολότερο με την BeautifulSoup είναι να θυμάστε πώς εγκαθίσταται και πώς εισάγεται στην Python.

Για να εγκαταστήσετε την BeautifulSoup, εκτελέστε την ακόλουθη εντολή στο τερματικό:

```
pip install beautifulsoup4
```

Μετά την εγκατάσταση, εισάγετε την BeautifulSoup ως εξής:

````python
from bs4 import BeautifulSoup
````

Αφού εισαγάγαμε την BeautifulSoup, μπορούμε να χρησιμοποιήσουμε την κλάση `BeautifulSoup`, η οποία μας επιτρέπει να αναλύουμε HTML. Όπως θα δούμε σε αυτό το κεφάλαιο, σπάνια θα έχετε την HTML ως συμβολοσειρά μέσα στο σενάριο Python· για αρχή, όμως, ας αναλύσουμε την παραπάνω συμβολοσειρά `html`, δίνοντάς τη στην κλάση `BeautifulSoup`. Συνηθίζεται (είναι Pythonic) να ονομάζετε τη μεταβλητή BeautifulSoup `soup`. Αν έχετε ένα πιο σύνθετο σενάριο που αναλύει αντικείμενα soup από πολλούς ιστότοπους, ίσως χρειαστεί να επιλέξετε πιο περιγραφικά ονόματα· για τους σκοπούς μας, όμως, το `soup` αρκεί.

````python
soup = BeautifulSoup(html)
````

Αφού δημιουργήσαμε το αντικείμενο `soup` στη μνήμη, μπορούμε να αρχίσουμε να το εξετάζουμε. Αν το εκτυπώσουμε, δεν θα παρατηρήσουμε κάτι ιδιαίτερο: μοιάζει με μια κανονική συμβολοσειρά.

````python
print(soup)
````

Μπορεί να μοιάζει με συμβολοσειρά, αλλά δεν είναι. Μπορούμε να το διαπιστώσουμε ρωτώντας την Python τι τύπου αντικείμενο είναι, με τη συνάρτηση `type`.

````python
type(soup)
````

Όπως βλέπουμε, πρόκειται για αντικείμενο της κλάσης `bs4.BeautifulSoup`. Αυτό σημαίνει ότι, παρότι μοιάζει με συμβολοσειρά, περιέχει στην πραγματικότητα περισσότερα δεδομένα, στα οποία μπορούμε να έχουμε πρόσβαση. Για παράδειγμα, με τη μέθοδο `.find` βρίσκουμε την πρώτη εμφάνιση μιας συγκεκριμένης ετικέτας. Η μέθοδος `find` έχει ένα υποχρεωτικό όρισμα: μια συμβολοσειρά με το όνομα της ετικέτας που θέλετε να εξαγάγετε. Ας ανακτήσουμε την πρώτη ετικέτα `div`.

````python
first_div = soup.find("div")
print(first_div)
````

Όπως βλέπετε, με την `.find` ανακτήσαμε την πρώτη ετικέτα `div`, αλλά γνωρίζουμε ότι η συμβολοσειρά HTML περιέχει περισσότερες. Τι κάνουμε αν θέλουμε να τις ανακτήσουμε όλες; Χρησιμοποιούμε τη μέθοδο `.find_all`. Όπως η `.find`, έτσι και η `.find_all` δέχεται ένα μόνο υποχρεωτικό όρισμα: πάλι τη συμβολοσειρά με το όνομα της ετικέτας που θέλετε να εξαγάγετε.

````python
all_divs = soup.find_all("div")
print(all_divs)
````

Σε αντίθεση με την `.find`, που επιστρέφει ένα μόνο στοιχείο, η `.find_all` επιστρέφει μια λίστα ετικετών. Τι γίνεται όμως αν δεν θέλουμε όλες τις ετικέτες, αλλά μόνο τις `div` με μια συγκεκριμένη τιμή στο χαρακτηριστικό `class`; Η BeautifulSoup μάς το επιτρέπει με ένα δεύτερο όρισμα στην `.find` ή στην `.find_all`: ένα λεξικό με κλειδιά τα χαρακτηριστικά και τιμές τις τιμές των χαρακτηριστικών των ετικετών που θέλετε να εξαγάγετε.

````python
div_other = soup.find_all("div", {"class": "other"})
print(div_other)
````

Αφού εντοπίσουμε την ετικέτα που θέλουμε να εξαγάγουμε από την HTML, μπορούμε να προσπελάσουμε τα ένθετα στοιχεία της. Κάθε τέτοιο αντικείμενο λειτουργεί όπως το αρχικό `soup` και περιέχει όλες τις θυγατρικές (children), δηλαδή ένθετες, ετικέτες. Μπορούμε επομένως να ανακτήσουμε την ετικέτα `p` που βρίσκεται μέσα στην ετικέτα `div` με τιμή `other` στο χαρακτηριστικό `class`, χρησιμοποιώντας το `.find("p")`.

````python
paragraph = div_other[0].find("p")
print(paragraph)
````

Όταν εργαζόμαστε με δεδομένα κειμένου, όμως, σπάνια θέλουμε να διατηρήσουμε την HTML· θέλουμε να εξαγάγουμε το κείμενο που περιέχει. Για τον σκοπό αυτό, το ακατέργαστο κείμενο μέσα στην HTML είναι προσβάσιμο με το `.text`.

````python
print(paragraph.text)
````

Επειδή όλα αυτά γίνονται προγραμματιστικά, μπορούμε να αυτοματοποιήσουμε το scraping αρχείων HTML με την Python. Μπορούμε, για παράδειγμα, να εξαγάγουμε όλο το κείμενο από κάθε ετικέτα `div` του αρχείου μας με τις ακόλουθες δύο γραμμές κώδικα.

````python
for div in all_divs:
    print(div.text)
````

## Πώς βρίσκουμε την HTML ενός ιστότοπου

Τώρα που έχετε μια γενική εικόνα της HTML και του τρόπου λειτουργίας της, ας δούμε την HTML ενός πραγματικού ιστότοπου. Στο επόμενο κεφάλαιο θα κάνουμε web scraping στη Wikipedia, οπότε ας εστιάσουμε σε αυτήν και εδώ. Αν ο φυλλομετρητής σας το υποστηρίζει (π.χ. Chrome και Firefox), μπορείτε να ενεργοποιήσετε τη λειτουργία προγραμματιστή (developer mode). Κάθε λειτουργικό σύστημα και φυλλομετρητής έχει διαφορετικές συντομεύσεις πληκτρολογίου γι' αυτό, αλλά σε όλους μπορείτε να κάνετε δεξί κλικ στην ιστοσελίδα και να επιλέξετε "Inspect" («Επιθεώρηση»), ώστε να ανοίξει η λειτουργία προγραμματιστή. Σε αυτό το σημείο του κεφαλαίου συνιστώ ιδιαίτερα να δείτε το βίντεο, γιατί θα σας είναι ευκολότερο να παρακολουθήσετε.

Σε αυτό το κεφάλαιο (και στο επόμενο) θα εργαστούμε με την εξής σελίδα:
https://en.wikipedia.org/wiki/List_of_French_monarchs

Ανοίξτε την σε μια άλλη οθόνη ή σε νέα καρτέλα. Το άρθρο αυτό της Wikipedia περιέχει λίγο κείμενο, αλλά κυρίως έναν κατάλογο όλων των μοναρχών της Γαλλίας, από τους Καρολίγγειους έως τον Ναπολέοντα Γ', στα μέσα του 19ου αιώνα.

Όταν επιθεωρείτε τη σελίδα, βλέπετε όλη την ένθετη HTML που περιέχει. Αφιερώστε λίγο χρόνο στην εξέταση των ετικετών. Στην επόμενη ενότητα θα μάθουμε πώς να κάνουμε scraping σε αυτή τη σελίδα· προς το παρόν, δείτε τι μπορούμε να κάνουμε με μερικές βασικές εντολές της Python.

````python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/List_of_French_monarchs"
s = requests.get(url)

soup = BeautifulSoup(s.content)
body = soup.find("div", {"id": "mw-content-text"})
for paragraph in body.find_all("p")[:5]:
    if paragraph.text.strip() != "":
        print (paragraph.text)
````

Στο παραπάνω κελί χρησιμοποιήσαμε δύο βιβλιοθήκες, τις requests και BeautifulSoup, για να στείλουμε αίτημα στον διακομιστή της Wikipedia που φιλοξενεί τη συγκεκριμένη σελίδα. Στη συνέχεια αναζητήσαμε την ετικέτα `div` που περιέχει το κύριο σώμα της σελίδας, δηλαδή την ετικέτα `div` με τιμή "mw-content-text" στο χαρακτηριστικό "id". Έπειτα αναζήτησα όλες τις ετικέτες "p" (παραγράφους) μέσα σε αυτό το σώμα και εκτύπωσα το κείμενο όσων δεν ήταν κενές. Στο τέλος του επόμενου κεφαλαίου όχι μόνο θα κατανοείτε τον παραπάνω κώδικα, αλλά θα μπορείτε να τον γράψετε και μόνοι σας. Προς το παρόν, επιθεωρήστε τη σελίδα και δείτε αν μπορείτε να εντοπίσετε στην HTML την ετικέτα `div` με id "mw-content-text". Δεν πειράζει αν δυσκολευτείτε! Δεν είναι κάτι που βγαίνει αυθόρμητα από την αρχή· χρειάζεται εξάσκηση. Ένα κόλπο για να ξεκινήσετε, πάντως, είναι να κάνετε δεξί κλικ στην περιοχή στην οποία θέλετε να κάνετε scraping και να επιλέξετε "Inspect".
